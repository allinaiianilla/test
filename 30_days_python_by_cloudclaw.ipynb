## Day 1：Python 变量与基本数据类型

**今日目标**：理解变量赋值、掌握 `int/float/str/bool` 四种最常用类型。

### 核心概念

## Day 2：字符串操作

**今日目标**：掌握字符串格式化、切片、常用方法。

### 核心概念
- f-string 是最推荐的格式化方式（Python 3.6+）
- 切片 `[start:end:step]` 是 Python 标志性操作
- `split/join/strip/replace` 是四大字符串方法

```python
# === Day 2 实操 ===

# 1. f-string — 最常用格式化
name, score = "Alice", 98.5
print(f"{name} 考了 {score} 分")

# 2. 切片 — Python 必杀技
s = "Hello Python"
print(s[0:5])    # Hello
print(s[:5])     # Hello（从头开始可省略0）
print(s[6:])     # Python（到最后可省略末尾）
print(s[::-1])   # nohtyP olleH（反转字符串！）

# 3. 四大字符串方法
raw = "  apple,banana ,cherry  "
print(raw.strip())              # 去首尾空格
print(raw.replace("banana", "orange"))  # 替换
print(raw.strip().split(","))  # 分割 → ['apple', 'banana ', 'cherry']

fruits = ["A", "B", "C"]
print("|".join(fruits))        # 拼接 → A|B|C

# 4. 判断方法
print("123".isdigit())  # True
print("abc".isalpha())  # True
print("Hello".startswith("He"))  # True
```

## Day 3：列表 — 最万能的数据容器

**今日目标**：列表的增删改查、遍历、排序。

### 核心概念
- 列表是**有序、可变**的容器
- 增：`append/extend/insert` | 删：`remove/pop`
- `enumerate()` 同时获取索引和值

```python
# === Day 3 实操 ===

# 1. 增删改查
items = [1, 2, 3]
items.append(4)      # [1,2,3,4] — 尾部加
items.insert(0, 0)   # [0,1,2,3,4] — 指定位置加
items.extend([5,6])  # [0,1,2,3,4,5,6] — 合并列表

items.remove(0)      # [1,2,3,4,5,6] — 按值删
last = items.pop()   # last=6, items=[1,2,3,4,5] — 弹尾
items[2] = 99        # [1,2,99,4,5] — 改值

print(items)
print(items.index(99))  # 2 — 查找位置
print(5 in items)       # True — 是否存在

# 2. 遍历（工作中最最常用的模式）
for i, val in enumerate(items):
    print(f"索引{i} → {val}")

# 3. 切片（列表同样适用）
print(items[1:3])   # [2, 99]
print(items[::-1])  # [5, 4, 99, 2, 1] — 反转

# 4. 排序
nums = [3, 1, 4, 1, 5, 9]
print(sorted(nums))           # [1,1,3,4,5,9] 不改变原列表
print(sorted(nums, reverse=True))  # 降序
nums.sort()                   # 原地排序
print(nums)
```

## Day 4：元组、集合、字典

**今日目标**：三大数据结构的区分与适用场景。

### 核心概念
| 类型 | 特点 | 场景 |
|------|------|------|
| tuple | 不可变 | 常量、函数多返回、字典键 |
| set | 无序、去重 | 去重、交并差运算 |
| dict | 键值对 | 配置、计数、缓存、JSON |

```python
# === Day 4 实操 ===

# 1. 元组 — 不可变的列表
point = (3, 4)
x, y = point   # 解包，超常用！
print(f"x={x}, y={y}")

# 2. 集合 — 去重神器
nums = [1, 2, 2, 3, 3, 3]
unique = list(set(nums))
print(unique)  # [1, 2, 3]

A = {1, 2, 3}
B = {3, 4, 5}
print(A & B)   # {3} — 交集
print(A | B)   # {1,2,3,4,5} — 并集
print(A - B)   # {1, 2} — 差集

# 3. 字典 — 80% 的 Python 代码都在用
user = {
    "name": "张三",
    "age": 28,
    "skills": ["Python", "SQL"]
}

# 安全取值（避免 KeyError）
print(user.get("email", "未设置"))  # 未设置

# 遍历字典
for key, value in user.items():
    print(f"{key} → {value}")

# 合并字典
extra = {"city": "北京", "age": 30}
merged = {**user, **extra}  # 后者覆盖前者
print(merged)

# 计数模式（面试常考）
words = ["a", "b", "a", "c", "b", "a"]
count = {}
for w in words:
    count[w] = count.get(w, 0) + 1
print(count)  # {'a': 3, 'b': 2, 'c': 1}
```

## Day 5：条件判断与布尔逻辑

**今日目标**：if/elif/else、短路求值、三元表达式。

### 核心概念
- Python 的缩进就是代码块，必须严格 4 空格

```python
# === Day 5 实操 ===

score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "D"

print(f"{score}分 → {grade}")

# 三元表达式（一行 if/else）
status = "通过" if score >= 60 else "未通过"

# 短路求值 — 工作中常见写法
name = None
display = name or "匿名用户"  # 取第一个真值
print(display)  # 匿名用户

# 链式比较（Python 独有，很 Pythonic）
age = 25
if 18 <= age <= 60:
    print("劳动年龄")

# 判空 — 空列表/空字符串/0/None 都是 False
items = []
if not items:
    print("列表为空")
```

## Day 6：for / while 循环

**今日目标**：掌握两种循环的场景区分、break/continue。

### 核心概念
- `for` 遍历可迭代对象（90% 的场景用它）
- `while` 条件循环（不确定次数的场景）
- `range()` 生成数字序列
- `break` 跳出循环，`continue` 跳过本次

```python
# === Day 6 实操 ===

# 1. for + range（遍历数字）
for i in range(5):        # 0,1,2,3,4
    print(i, end=" ")
print()

for i in range(2, 6):     # 2,3,4,5
    print(i, end=" ")
print()

# 2. break — 找到就停（效率优化关键）
data = [1, 3, 5, 7, -1, 9]
for d in data:
    if d < 0:
        print(f"发现负数: {d}")
        break

# 3. continue — 跳过不符合条件的
for n in range(10):
    if n % 2 == 0:
        continue  # 跳过偶数
    print(n, end=" ")  # 1 3 5 7 9
print()

# 4. while — 读文件/轮询等场景
count = 0
while count < 3:
    print(f"第{count+1}次尝试")
    count += 1

# 5. for + else（Python 特有）— 循环正常结束才执行
for x in [1, 2, 3]:
    if x > 10:
        break
else:
    print("没找到大于10的数（for循环正常结束）")
```

## Day 7：列表/字典推导式 — Python 的效率之魂

**今日目标**：一行代码替代多行循环，Python 程序员必备技能。

### 核心概念
- 列表推导式：`[表达式 for 变量 in 可迭代对象 if 条件]`
- 字典推导式：`{key: value for ...}`
- 集合推导式：`{表达式 for ...}`

```python
# === Day 7 实操 ===

# 1. 列表推导式 — 三行变一行
# 传统写法
sq1 = []
for i in range(10):
    if i % 2 == 0:
        sq1.append(i ** 2)

# 推导式写法
sq2 = [i**2 for i in range(10) if i % 2 == 0]
print(sq2)  # [0, 4, 16, 36, 64]

# 2. 字典推导式 — 日常超高频
names = ["Alice", "Bob", "Charlie"]
name_len = {name: len(name) for name in names}
print(name_len)  # {'Alice': 5, 'Bob': 3, 'Charlie': 7}

# 翻转字典
d = {"a": 1, "b": 2, "c": 3}
flipped = {v: k for k, v in d.items()}
print(flipped)  # {1: 'a', 2: 'b', 3: 'c'}

# 3. 带条件过滤
nums = [1, 2, 3, 4, 5, 6]
evens = [n for n in nums if n % 2 == 0]
print(evens)

# 4. 嵌套推导式（如展平二维列表）
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [x for row in matrix for x in row]
print(flat)  # [1, 2, 3, 4, 5, 6]

# 5. 生成器表达式 — 省内存
# 把方括号换成圆括号即可
big_sum = sum(i**2 for i in range(1000000))  # 不会生成 100 万的中间列表
print(f"平方和: {big_sum}")
```

## Day 8：zip / enumerate / sorted — 最常用的三个内置函数

**今日目标**：三个函数搞定日常数据处理。

```python
# === Day 8 实操 ===

# 1. enumerate — 遍历时带索引
fruits = ["苹果", "香蕉", "橘子"]
for i, fruit in enumerate(fruits):
    print(f"#{i+1} {fruit}")

# 2. zip — 同时遍历多个列表
names = ["张三", "李四", "王五"]
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f"{name}: {score}分")

# 拆 zip（解压）
paired = list(zip(names, scores))
unzip_names, unzip_scores = zip(*paired)
print(unzip_names)   # ('张三', '李四', '王五')
print(unzip_scores)  # (85, 92, 78)

# 3. sorted 排序
students = [
    {"name": "张三", "score": 85},
    {"name": "李四", "score": 92},
    {"name": "王五", "score": 78},
]
# 按 score 降序
ranked = sorted(students, key=lambda s: s["score"], reverse=True)
for s in ranked:
    print(f"{s['name']}: {s['score']}")

# 4. any / all
print(any([False, False, True]))  # True（有真即可）
print(all([True, True, False]))   # False（全真才行）
print(all(80 <= s["score"] <= 100 for s in students))  # 所有人成绩>=80?
```

## Day 9：函数基础 — 定义、参数、返回值

**今日目标**：写好自己的函数，理解参数类型。

```python
# === Day 9 实操 ===

# 1. 基本函数
def greet(name):
    """打招呼函数 — 这是文档字符串（docstring）"""
    return f"你好, {name}!"

print(greet("张三"))

# 2. 默认参数 — 工作中最常用的函数写法
def create_user(name, age=18, city="北京"):
    return {"name": name, "age": age, "city": city}

print(create_user("张三"))                 # 全部默认
print(create_user("李四", age=25))          # 指定部分
print(create_user("王五", city="上海"))     # 关键字参数

# ⚠️ 陷阱：默认参数别用可变对象！
# 错误示范: def f(lst=[]): ...  ← 不要这样写
# 正确写法:
def add_item(item, lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst

# 3. *args 和 **kwargs（处理不定数量参数）
def log_info(*args, **kwargs):
    print("位置参数:", args)
    print("关键字参数:", kwargs)

log_info("error", "api_fail", user="张三", code=500)

# 4. 多返回值（实际是返回元组）
def min_max(nums):
    return min(nums), max(nums)

lo, hi = min_max([3, 1, 4, 1, 5])
print(f"最小:{lo}, 最大:{hi}")
```

## Day 10：lambda 与高阶函数

**今日目标**：理解 lambda、map/filter/reduce、闭包基础。

### 核心概念
- lambda 就是**一次性匿名函数**
- 排序 `key=`、`map/filter`、pandas `apply` 是最主要的 lambda 场景
- 闭包：函数里套函数，内层记住外层变量

```python
# === Day 10 实操 ===

# 1. lambda — 短小精悍
add = lambda x, y: x + y
print(add(3, 5))  # 8

# 排序花样（lambda 最常见使用场景）
pairs = [(1, "b"), (2, "a"), (3, "c")]
print(sorted(pairs, key=lambda p: p[1]))          # 按第二个元素排
print(sorted(pairs, key=lambda p: p[0], reverse=True))  # 降序

# 2. map — 对每个元素做同一操作
nums = [1, 2, 3, 4, 5]
print(list(map(lambda x: x**2, nums)))  # [1, 4, 9, 16, 25]
# 但推导式更 Pythonic: [x**2 for x in nums]

# 3. filter — 过滤
print(list(filter(lambda x: x > 3, nums)))  # [4, 5]
# 推导式更 Pythonic: [x for x in nums if x > 3]

# 4. 闭包 — 函数工厂
def make_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(5))  # 10
print(triple(5))  # 15
```

## Day 11：函数进阶 — 装饰器、类型注解

**今日目标**：理解装饰器本质、会写简单装饰器、了解类型注解。

### 核心概念
- 装饰器 = 给函数包一层，不改变原函数，加额外功能
- `@timer` 就是 `func = timer(func)` 的语法糖
- 类型注解让代码更易读，IDE 更智能

```python
# === Day 11 实操 ===

import time
from functools import wraps
from typing import Callable, List, Dict

# 1. 计时装饰器 — 最实用的装饰器
def timer(func: Callable) -> Callable:
    @wraps(func)  # 保留原函数元信息
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        elapsed = time.time() - start
        print(f"[{func.__name__}] 耗时: {elapsed:.4f}秒")
        return result
    return wrapper

@timer
def slow_sum(n):
    time.sleep(0.1)
    return sum(range(n))

print(slow_sum(1000000))

# 2. 重试装饰器
def retry(max_attempts=3):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    if attempt == max_attempts:
                        raise
                    print(f"  重试 {attempt}/{max_attempts}: {e}")
                    time.sleep(0.5)
        return wrapper
    return decorator

# 3. 类型注解 — 让代码自带文档
def calculate_average(scores: List[float]) -> float:
    """计算平均分"""
    return sum(scores) / len(scores) if scores else 0.0

result: float = calculate_average([85.5, 92.0, 78.5])
print(f"平均分: {result:.1f}")
```

## Day 12：异常处理

**今日目标**：写出健壮的代码，不乱崩。

### 核心概念
- `try/except/finally` 是必备技能
- **捕获具体异常**，不要裸 `except:`
- `finally` 做清理，`else` 无异常时执行
- 自定义异常让错误更清晰

```python
# === Day 12 实操 ===

# 1. 基本异常处理
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print(f"错误: 除数不能为零 (a={a}, b={b})")
        return None
    except TypeError as e:
        print(f"类型错误: {e}")
        return None
    finally:
        # 清理操作（如关闭文件、释放锁等）

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # None

# 2. 实际场景：解析用户输入
def parse_age(user_input):
    try:
        age = int(user_input)
        if age < 0 or age > 150:
            raise ValueError(f"年龄不合理: {age}")
        return age
    except ValueError as e:
        print(f"输入错误: {e}")
        return None

print(parse_age("25"))      # 25
print(parse_age("abc"))     # None
print(parse_age("200"))     # None

# 3. 自定义异常
class ConfigError(Exception):
    """配置相关错误"""
    pass

def load_config(key):
    config = {"host": "localhost"}
    if key not in config:
        raise ConfigError(f"缺少配置项: {key}")
    return config[key]

try:
    load_config("port")
except ConfigError as e:
    print(f"配置错误: {e}")

# 4. 上下文管理器模式（with 语句）
# with 自动调用 __enter__ / __exit__，确保资源释放
from contextlib import contextmanager

@contextmanager
def timer_context(name="操作"):
    start = time.time()
    try:
        yield  # 这里执行 with 块里的代码
    finally:
        print(f"[{name}] 耗时: {time.time()-start:.3f}秒")

with timer_context("数据处理"):
    total = sum(range(1000000))
    print(f"结果: {total}")
```

```python
# === Day 14 实操 ===

import json
from datetime import datetime, date

# 1. 基本转换：dict ↔ JSON
data = {
    "name": "张三",
    "age": 28,
    "skills": ["Python", "SQL", "Docker"],
    "active": True,
    "score": None
}

# Python → JSON 字符串
json_str = json.dumps(data, ensure_ascii=False, indent=2)
print("JSON:")
print(json_str)

# JSON 字符串 → Python
parsed = json.loads(json_str)
print(f"\n解析: {parsed['name']}")

# 2. 读写 JSON 文件
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

with open("data.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)
    print(f"从文件读取: {loaded['skills']}")

# 3. 处理 datetime（JSON 不认识）
class DateTimeEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, (datetime, date)):
            return obj.isoformat()
        return super().default(obj)

log = {"event": "login", "time": datetime.now()}
print(json.dumps(log, cls=DateTimeEncoder, ensure_ascii=False))

# 清理
os.remove("data.json")
```

## Day 19：正则表达式

**今日目标**：掌握最常用的 10 个正则模式，文本处理利器。

### 核心概念
- 正则就是**模式匹配语言**
- `re.search/findall/sub` 是三巨头
- **不用背**，忘了查表就行，但理解基本规则

```python
# === Day 19 实操 ===

import re

print("=== 10 个最常用的正则模式 ===\n")

# 1. 提取数字
text = "订单号: ORD-2024-00123, 金额: ¥199.99"
nums = re.findall(r'\d+\.?\d*', text)
print(f"1. 提取数字: {nums}")

# 2. 提取邮箱
text2 = "联系: alice@example.com 和 bob@test.cn"
emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text2)
print(f"2. 提取邮箱: {emails}")

# 3. 提取手机号
text3 = "手机: 13800138000, 座机: 010-12345678"
phones = re.findall(r'1[3-9]\d{9}', text3)
print(f"3. 提取手机号: {phones}")

# 4. 提取中文
text4 = "Python 是一种编程语言，release 于 1991 年"
chinese = re.findall(r'[\u4e00-\u9fff]+', text4)
print(f"4. 提取中文: {chinese}")

# 5. 替换 — sub
text5 = "密码是 123456，千万别泄露"
masked = re.sub(r'\d+', '****', text5)
print(f"5. 脱敏替换: {masked}")

# 6. 分割
text6 = "张三,李四;王五|赵六"
parts = re.split(r'[,;|]', text6)
print(f"6. 多符号分割: {parts}")

# 7. search — 找到第一个匹配
match = re.search(r'\d+', "价格是 199 元")
if match:
    print(f"7. search: {match.group()} 在位置 {match.start()}")

# 8. 命名分组
log = "2024-01-15 10:30:45 ERROR 连接超时"
m = re.search(
    r'(?P<date>\d{4}-\d{2}-\d{2}) (?P<time>\d{2}:\d{2}:\d{2}) (?P<level>\w+) (?P<msg>.+)',
    log
)
if m:
    print(f"8. 命名分组: {m.groupdict()}")

# 9. 贪婪 vs 非贪婪
html = "<title>我的网页</title><body>内容</body>"
print(f"9a. 贪婪: {re.findall(r'<.*>', html)}")
print(f"9b. 非贪婪: {re.findall(r'<.*?>', html)}")

# 10. 预编译（高频匹配时提升性能）
phone_pattern = re.compile(r'1[3-9]\d{9}')
print(f"10. 预编译匹配: {phone_pattern.findall('13800138000, 13900139000')}")

print("\n常用速查: \\d=数字 \\w=字母数字 \\s=空白 .*=任意次数 +=1次+ ?=可选 [] =字符组 ()=分组")
```

## Day 21：面向对象基础 — 类与实例

**今日目标**：类的基本语法、`__init__`、实例方法。

### 核心概念
- 类 = 蓝图，实例 = 按蓝图造出来的对象
- `self` 代表实例本身，是 Python 的约定（不是关键字）
- 属性 + 方法 = 数据 + 行为

```python
# === Day 21 实操 ===

class User:
    """用户类 — 最基础的 OOP 例子"""

    # 类属性（所有实例共享）
    platform = "Web"

    def __init__(self, name, age, email=None):
        """构造方法 — 创建实例时自动调用"""
        self.name = name      # 实例属性
        self.age = age
        self.email = email or f"{name}@example.com"
        self._login_count = 0  # 下划线开头 = 约定私有

    def login(self):
        """实例方法"""
        self._login_count += 1
        return f"{self.name} 第{self._login_count}次登录"

    def __str__(self):
        """打印对象的友好表示"""
        return f"User({self.name}, {self.age}岁)"

    def __repr__(self):
        """调试表示"""
        return f"User(name='{self.name}', age={self.age})"

# 使用
u1 = User("张三", 28)
u2 = User("李四", 32, "lisi@qq.com")

print(u1)                    # User(张三, 28岁)
print(repr(u1))             # User(name='张三', age=28)
print(u1.login())           # 张三 第1次登录
print(u1.login())           # 张三 第2次登录
print(f"邮箱: {u1.email}")  # 张三@example.com
print(f"邮箱: {u2.email}")  # lisi@qq.com
print(f"平台: {User.platform}")  # 类属性
```

## Day 22：继承与多态

**今日目标**：父子类关系、方法重写、`super()`。

```python
# === Day 22 实操 ===

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} 发出声音"

    def move(self):
        return "移动"


class Dog(Animal):
    def speak(self):          # 方法重写
        return f"{self.name}: 汪汪!"

    def fetch(self):          # 子类特有方法
        return "叼回骨头"


class Cat(Animal):
    def speak(self):
        return f"{self.name}: 喵喵~"


# 多态 — 同一接口，不同行为
animals = [Dog("旺财"), Cat("小花"), Dog("阿黄")]
for a in animals:
    print(a.speak())

# super() — 调父类方法
class RobotDog(Dog):
    def __init__(self, name, battery):
        super().__init__(name)  # 调父类 __init__
        self.battery = battery

    def speak(self):
        parent_msg = super().speak()  # 调父类 speak
        return f"🤖 {parent_msg} (电量:{self.battery}%)"


robot = RobotDog("铁蛋", 80)
print(robot.speak())

# isinstance 检查继承关系
print(f"robot 是 Dog? {isinstance(robot, Dog)}")       # True
print(f"robot 是 Animal? {isinstance(robot, Animal)}") # True (间接继承)
```

## Day 23：魔术方法与属性

**今日目标**：`__len__/__getitem__/__iter__`、`@property`。

### 核心概念
- 魔术方法让你的类像内置类型一样工作
- `@property` 把方法变成属性访问

```python
# === Day 23 实操 ===

class Playlist:
    def __init__(self, name, songs=None):
        self.name = name
        self._songs = songs or []

    # 让对象支持 len()
    def __len__(self):
        return len(self._songs)

    # 让对象支持 [] 索引
    def __getitem__(self, index):
        return self._songs[index]

    # 让对象可迭代
    def __iter__(self):
        return iter(self._songs)

    # 让对象支持 in 操作
    def __contains__(self, item):
        return item in self._songs

    # 让对象支持 +
    def __add__(self, other):
        return Playlist(
            f"{self.name}+{other.name}",
            self._songs + other._songs
        )


pl = Playlist("最爱", ["晴天", "七里香", "夜曲"])
print(f"长度: {len(pl)}")
print(f"第一首: {pl[0]}")
print("\"夜曲\" 在列表里?", "夜曲" in pl)

# 合并
pl2 = Playlist("新歌", ["告白气球"])
merged = pl + pl2
print(f"合并: {list(merged)}")

# === @property — getter/setter 更优雅 ===

class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price

    @property
    def price(self):
        """像属性一样访问"""
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("价格不能为负")
        self._price = value

    @property
    def price_with_tax(self):
        """计算属性"""
        return round(self._price * 1.13, 2)


p = Product("《Python入门》", 59.0)
print(f"价格: ¥{p.price}, 含税: ¥{p.price_with_tax}")
p.price = 69.0  # 通过 setter 赋值
print(f"新价格: ¥{p.price}")
```

## Day 24：collections — 进阶数据结构

**今日目标**：Counter、defaultdict、OrderedDict、namedtuple。

```python
# === Day 24 实操 ===

from collections import Counter, defaultdict, OrderedDict, namedtuple

# 1. Counter — 计数器（统计频率神器）
words = list("abracadabra")
c = Counter(words)
print(f"字母频率: {c}")
print(f"最常见2个: {c.most_common(2)}")

# 2. defaultdict — 有默认值的字典
# 不用再写: d[key] = d.get(key, []) + ...
groups = defaultdict(list)
data = [("A", 1), ("B", 2), ("A", 3), ("C", 4), ("B", 5)]
for key, val in data:
    groups[key].append(val)
print(f"分组: {dict(groups)}")

# 默认数量
counts = defaultdict(int)
for w in ["a", "b", "a", "c", "b", "a"]:
    counts[w] += 1  # 不用先判断 key 是否存在！
print(f"计数: {dict(counts)}")

# 3. namedtuple — 像字典一样可读，像元组一样轻量
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(f"x={p.x}, y={p.y}")
print(f"距离原点: {(p.x**2 + p.y**2)**0.5}")

# 4. OrderedDict — 保持插入顺序（Python 3.7+ 普通 dict 也保序了）
od = OrderedDict()
od["first"] = 1
od["second"] = 2
print(f"有序字典: {list(od.keys())}")
```

## Day 25：itertools — 迭代器工具箱

**今日目标**：链式迭代、排列组合、无限迭代、分组。

```python
# === Day 25 实操 ===

from itertools import chain, combinations, groupby, islice, product

# 1. chain — 合并多个可迭代对象
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = [7, 8]
print(f"合并: {list(chain(list1, list2, list3))}")

# 2. combinations — 组合
print(f"C(4,2) 所有组合: {list(combinations('ABCD', 2))}")

# 3. 笛卡尔积
print(f"骰子×硬币: {list(product([1,2,3,4,5,6], ['正','反']))[:8]}...")

# 4. groupby — 连续相同元素分组
items = sorted([("水果", "苹果"), ("水果", "香蕉"), ("蔬菜", "白菜")])
for key, group in groupby(items, key=lambda x: x[0]):
    print(f"{key}: {[x[1] for x in group]}")

# 5. islice — 迭代器切片
print(f"islice切片: {list(islice(range(100), 1, 7, 2))}")  # 索引1~6步长2 → [1, 3, 5]

print("\n其他: permutations=排列, zip_longest=长短zip")
```

## Day 27：requests — HTTP 客户端

**今日目标**：GET/POST 请求、请求头、JSON、错误处理。

```python
# === Day 27 实操 ===

# 如果没装 requests: !pip install requests

print("=== requests 核心模式 ===\n")

import requests

# 1. GET 请求（使用 httpbin.org 安全测试站点）
r = requests.get("https://httpbin.org/get?name=test", timeout=10)
print(f"状态码: {r.status_code}")
data = r.json()
print(f"参数: {data.get('args', {})}")

# 2. POST 请求
payload = {"name": "张三", "age": 28}
r = requests.post("https://httpbin.org/post", json=payload, timeout=10)
print(f"\nPOST 响应: {r.json().get('json', {})}")

# 3. 带请求头
headers = {"User-Agent": "Python-Learner/1.0"}
r = requests.get("https://httpbin.org/headers", headers=headers, timeout=10)
print(f"\n自定义头: {r.json().get('headers', {}).get('User-Agent')}")

# 4. 错误处理模板
url = "https://httpbin.org/status/404"
try:
    r = requests.get(url, timeout=10)
    r.raise_for_status()  # 非 2xx 自动抛异常
except requests.Timeout:
    print("请求超时")
except requests.HTTPError as e:
    print(f"HTTP 错误: {e}")
except requests.RequestException as e:
    print(f"请求失败: {e}")

# 5. Session — 复用连接
s = requests.Session()
s.headers.update({"User-Agent": "MyApp/1.0"})
r1 = s.get("https://httpbin.org/cookies/set?key=value", timeout=10)
r2 = s.get("https://httpbin.org/cookies", timeout=10)
print(f"\nSession Cookies: {r2.json().get('cookies', {})}")

print("\n🔗 状态码: 200=成功 201=创建 400=请求错误 401=未授权 403=禁止 404=不存在 500=服务器错误")
```

## Day 29：实战 — 简易网页爬虫

**今日目标**：综合运用 requests + BeautifulSoup + JSON 写一个小爬虫。

> ⚠️ 需安装: `pip install requests beautifulsoup4`

```python
# === Day 29 实操 ===

try:
    import requests
    from bs4 import BeautifulSoup
    import json
    import time

    # 1. 获取网页
    url = "https://httpbin.org/html"
    r = requests.get(url, timeout=10)
    r.raise_for_status()

    # 2. 解析 HTML
    soup = BeautifulSoup(r.text, "html.parser")
    title = soup.title.string if soup.title else "无标题"
    body = soup.body.get_text(strip=True)[:200] if soup.body else "无内容"
    print(f"标题: {title}")
    print(f"正文(前200字符): {body}")

    # 3. 爬虫模板
    print("\n=== 通用爬虫模板 ===")
    print("""
    import requests
    from bs4 import BeautifulSoup

    def scrape_page(url):
        headers = {"User-Agent": "Mozilla/5.0"}
        r = requests.get(url, headers=headers, timeout=10)
        r.raise_for_status()
        soup = BeautifulSoup(r.text, 'html.parser')
        # 这里写你的提取逻辑
        return soup

    # 翻页爬取
    results = []
    for page in range(1, 6):
        url = f"https://example.com/list?page={page}"
        soup = scrape_page(url)
        # 提取数据...
        time.sleep(1)  # 礼貌延迟

    # 保存
    with open('results.json', 'w') as f:
        json.dump(results, f, ensure_ascii=False, indent=2)
    """)

except ImportError:
    print("缺少依赖，请运行:")
    print("  !pip install requests beautifulsoup4")
except Exception as e:
    print(f"请求失败: {e}")
    print("(如果是网络受限环境，这属于正常情况)")
```

## Day 30：实战 — 数据分析小项目

**今日目标**：综合运用 pandas + matplotlib 做一个完整的数据分析。

> ⚠️ 需安装: `pip install pandas matplotlib`

```python
# === Day 30 实操 ===

# 如果没有 pandas/matplotlib: !pip install pandas matplotlib

print("=== 完整数据分析流程 ===\n")

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# 设置中文字体（避免中文乱码）
plt.rcParams["font.sans-serif"] = ["SimHei", "Arial Unicode MS", "DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

# ===== 1. 模拟创建销售数据 =====
df = pd.DataFrame({
    "订单ID": [f"ORD{i:04d}" for i in range(1, 21)],
    "日期": pd.date_range("2024-01-01", periods=20, freq="5D"),
    "产品": ["笔记本", "鼠标", "键盘", "显示器", "耳机"] * 4,
    "金额": [5999, 199, 499, 2499, 899,
             6299, 169, 529, 2199, 799,
             5499, 249, 459, 2699, 999,
             6799, 189, 599, 2399, 849],
    "数量": [1, 3, 2, 1, 2,
             1, 4, 2, 1, 3,
             1, 2, 2, 1, 1,
             1, 3, 1, 1, 2],
})
print(f"原始数据: {len(df)} 行")
print(df.head())

# ===== 2. 数据清洗 =====
df["日期"] = pd.to_datetime(df["日期"])
df["月份"] = df["日期"].dt.month
df["销售额"] = df["金额"] * df["数量"]
print(f"\n缺失值:\n{df.isnull().sum()}")

# ===== 3. 数据分析 =====
# 按月统计
monthly = df.groupby("月份").agg(
    销售额=("销售额", "sum"),
    订单数=("订单ID", "count"),
    平均单价=("金额", "mean"),
).round(2)
print(f"\n月度汇总:\n{monthly}")

# 按产品统计 Top5
top_products = df.groupby("产品")["销售额"].sum().nlargest(5)
print(f"\nTop5 热销产品:\n{top_products}")

# ===== 4. 可视化 =====
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 月度趋势
monthly["销售额"].plot(kind="bar", ax=axes[0], color="steelblue")
axes[0].set_title("月度销售额趋势")
axes[0].set_ylabel("销售额（元）")

# 产品占比
top_products.plot(kind="pie", autopct="%1.1f%%", ax=axes[1])
axes[1].set_title("Top5 产品占比")
axes[1].set_ylabel("")

plt.tight_layout()
plt.savefig("sales_report.png", dpi=150)
print("\n✅ 报告已保存: sales_report.png")

# ===== 5. 导出结果 =====
try:
    with pd.ExcelWriter("sales_report.xlsx") as writer:
        monthly.to_excel(writer, sheet_name="月度汇总")
        top_products.to_excel(writer, sheet_name="产品Top5")
    print("✅ Excel 报告已保存: sales_report.xlsx")
except ImportError:
    print("⚠️ 需要 openpyxl 才能导出 Excel: pip install openpyxl")

print("\n" + "=" * 50)
print("🎉 恭喜完成 30 天 Python 核心精要！")
print("=" * 50)
print("""
接下来你该做的事：
1. 自己动手写点小项目（爬虫、数据分析、自动化脚本）
2. 遇到问题先去 Stack Overflow 搜
3. 看别人代码学习（GitHub 上搜你感兴趣的项目）
4. 不需要记住所有 API，但要知道「大概有什么」能实现
5. 持续练习，编程是手艺活儿
""")
```
