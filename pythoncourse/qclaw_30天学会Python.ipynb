## Day 1：变量与基本类型

Python 是动态类型语言：变量不需要声明类型，赋值即创建。

最常用的基本类型：
- `int` 整数：`18`
- `float` 浮点数：`9.9`
- `str` 字符串：`"小明"`
- `bool` 布尔：`True / False`
- `NoneType` 空值：`None`（表示"没有值"）
- 判断 `None` 用 `is None`（身份比较），不要用 `== None`

常用内置：`type()` 看类型，`int()/str()/float()` 做类型转换。

```python
# 变量赋值（无需声明类型）
age = 18           # int
price = 9.9        # float
name = "小明"       # str
is_student = True  # bool
score = None       # None 表示空值

print(age, price, name, is_student, score)
print(type(age), type(price), type(name))

# 类型转换
n = int("42")      # 字符串 -> 整数
s = str(3.14)      # 数字 -> 字符串
print(n, s, type(n), type(s))

# 判断 None 用 is（身份比较），不用 ==
value = None
print(value is None, value is not None)   # True False
```

## Day 2：字符串：拼接、f-string、切片、常用方法

字符串用单/双/三引号包裹。日常最高频操作：

- 拼接：`+` 或 f-string
- **f-string**（最推荐）：`f"你好，{name}"`
- 切片：`s[起点:终点:步长]`
- 常用方法：`strip()` 去空格、`upper()/lower()` 大小写、`split()` 分割、`join()` 连接、`replace()` 替换、`startswith()/endswith()` 判断前缀后缀

```python
text = "  Hello, Python  "
print(text.strip())                 # 去两端空格
print(text.upper())                 # 转大写
print("hi" + " " + "there")         # 拼接

name = "小明"
print(f"你好，{name}，你学了 {3} 天")   # f-string（最常用）

s = "abcdef"
print(s[0], s[-1], s[1:4], s[::2])   # 切片：起点:终点:步长

words = "a,b,c".split(",")            # 分割
print(words)
print("-".join(words))                # 连接
print("Python".startswith("Py"), "a,b,c".replace(",", "|"))
```

## Day 3：列表 list：最常用的数据结构

列表是有序、可变的容器，几乎每个程序都会用到。

- 增：`append()` 末尾追加、`insert()` 指定位置插入
- 删：`remove()` 按值删、`pop()` 弹出末尾
- 改：用索引 `nums[i] = x`
- 查：`in` 判断存在、`index()` 找位置、`len()` 长度
- 遍历：`for x in nums:`
- 复制：`lst.copy()` 或 `lst[:]`（避免两个变量指向同一份数据）

```python
nums = [1, 2, 3]
nums.append(4)         # 末尾追加
nums.insert(0, 0)      # 指定位置插入
print(nums)

nums[1] = 20           # 修改
nums.remove(20)        # 按值删除
last = nums.pop()      # 弹出末尾
print(nums, last)

print(3 in nums, nums.index(3), len(nums))

for x in nums:
    print(x)

# 复制列表（避免共享同一份数据）
copy_nums = nums.copy()
copy_nums.append(99)
print(nums, copy_nums)   # nums 不变，copy_nums 多了 99
```

## Day 4：字典 dict：键值对，查找极快

字典用 `键: 值` 存储，按 key 取值速度极快（O(1)），适合表示"有名字的字段"（如一条记录）。

- 取值：`d[key]` 或 `d.get(key, 默认值)`（推荐 get，避免 KeyError）
- 增改：`d[key] = value`
- 删：`del d[key]`
- 遍历：`for k, v in d.items():`

```python
user = {"name": "小明", "age": 18, "city": "北京"}
print(user["name"])
print(user.get("email", "未填写"))    # 安全取值，键不存在返回默认值

user["age"] = 19                      # 修改
user["email"] = "a@b.com"             # 新增
del user["city"]                      # 删除
print(user)

for k, v in user.items():             # 遍历键值
    print(k, "=", v)
```

## Day 5：元组 tuple 与 集合 set

**元组 tuple**：不可变列表，适合"固定结构"的数据（坐标、函数返回多个值）。
**集合 set**：元素不重复，自动去重；支持交集 `&`、并集 `|` 等运算。

小技巧：函数 `return a, b` 返回的其实就是元组，可用 `x, y = func()` 解包。

```python
point = (3, 4)            # 元组，创建后不可修改
x, y = point              # 解包
print(x, y)

def min_max(values):
    return min(values), max(values)   # 返回多个值 = 返回元组
lo, hi = min_max([1, 9, 3])
print(lo, hi)

s = {1, 2, 2, 3, 3}       # 集合自动去重
print(s)
print({1, 2, 3} & {2, 3, 4})   # 交集
print({1, 2, 3} | {3, 4, 5})   # 并集
```

## Day 6：条件判断 if / elif / else

用缩进（通常 4 个空格）表示代码块，这是 Python 的硬性语法。

- 比较：`== != > < >= <=`
- 逻辑：`and` `or` `not`
- 三元表达式：`a if 条件 else b`（一行写简单分支）

```python
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "D"
print(grade)

# 三元表达式（简洁版 if/else）
status = "成年" if score >= 18 else "未成年"
print(status)
```

## Day 7：while 循环

当条件为真时重复执行。**务必保证循环能结束**，否则会死循环。

- `break` 立即结束整个循环
- `continue` 跳过本次，进入下一轮

```python
count = 0
while count < 5:
    print(count)
    count += 1

# break / continue 示例
n = 0
while True:
    n += 1
    if n % 2 == 0:
        continue       # 跳过偶数
    if n > 7:
        break          # 结束循环
    print(n)
```

## Day 8：for 循环 与 range

遍历序列（列表、字符串、字典…）首选 `for`。`range()` 生成整数序列。

- `range(5)` → 0,1,2,3,4
- `range(1, 10, 2)` → 1,3,5,7,9（起点, 终点(不含), 步长）
- `enumerate()`：同时拿到索引和值（遍历列表时几乎必用）

```python
for i in range(5):             # 0..4
    print(i)

for i in range(1, 10, 2):      # 1,3,5,7,9
    print(i)

fruits = ["苹果", "香蕉", "橙子"]
for idx, f in enumerate(fruits):    # 带索引遍历
    print(idx, f)
```

## Day 9：列表推导式：最 Pythonic 的技巧

一行生成列表，可读时优先用它代替 `for + append`。

语法：`[表达式 for 变量 in 可迭代对象 if 条件]`
可读性第一：嵌套超过两层就建议改回普通循环。

```python
squares = [x * x for x in range(10)]           # 平方
print(squares)

evens = [x for x in range(20) if x % 2 == 0]       # 加条件：只取偶数
print(evens)

words = ["Hello", "World"]
lower = [w.lower() for w in words]
print(lower)

# 等价传统写法（推导式更简洁）：
# evens = []
# for x in range(20):
#     if x % 2 == 0:
#         evens.append(x)
```

## Day 10：函数：定义、参数、返回值

函数让代码可复用、可测试。`def` 定义，`return` 返回结果。

- 位置参数：按顺序传
- 关键字参数：`greet("小红", greeting="早上好")`
- 默认参数：`def greet(name, greeting="你好")`

## Day 11：函数进阶：默认参数、*args、**kwargs、lambda

- `*args`：接收任意个位置参数（打包成元组）
- `**kwargs`：接收任意个关键字参数（打包成字典）
- `lambda`：一行匿名小函数，常用于 `key=` 排序等场景

⚠️ 经典坑：默认参数只在函数**定义时**求值一次。用可变对象（如列表）做默认值，会在多次调用间共享同一对象。正确做法：默认用 `None`，在函数内初始化。

```python
def f(a, b=10, *args, **kwargs):
    print("a=", a, "b=", b, "args=", args, "kwargs=", kwargs)

f(1, 2, 3, 4, x=5, y=6)

# lambda：一行小函数
square = lambda x: x * x
print(square(4))

nums = [5, 2, 9, 1]
nums.sort(key=lambda v: -v)      # 按相反顺序排序
print(nums)

# ⚠️ 可变默认参数的陷阱
def collect(x, bag=[]):        # 错误：bag 在多次调用间共享
    bag.append(x)
    return bag
print(collect(1), collect(2))  # [1, 2] 叠加，通常不是想要的

def collect_ok(x, bag=None):   # 正确：默认用 None，内部初始化
    bag = bag or []
    bag.append(x)
    return bag
print(collect_ok(1), collect_ok(2))  # [1] [2]
```

## Day 12：异常处理 try / except

用 `try/except` 捕获错误，避免程序直接崩溃。

- `else`：没出错时执行
- `finally`：无论是否出错都执行（常用于关闭资源）
常见异常：`ValueError` `KeyError` `FileNotFoundError` `ZeroDivisionError`

```python
try:
    n = int("abc")               # 这行会抛 ValueError
except ValueError as e:
    print("转换失败：", e)
else:
    print("转换成功：", n)
finally:
    print("无论如何都会执行")

# 更常见的实用写法
def safe_div(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None

print(safe_div(1, 0))
```

## Day 13：文件读写 open / with

用 `with open(...) as f` 能自动关闭文件，最安全。

- 模式：`"r"` 读、`"w"` 写（覆盖）、`"a"` 追加
- **读写中文务必加 `encoding="utf-8"`**
- 逐行读：`for line in f:`

```python
# 写文件
with open("demo.txt", "w", encoding="utf-8") as f:
    f.write("第一行\n")
    f.write("第二行\n")

# 读全部内容
with open("demo.txt", "r", encoding="utf-8") as f:
    content = f.read()
print(content)

# 逐行读取
with open("demo.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(">>>", line.strip())
```

## Day 15：高频内置函数（几乎每天都会用）

## Day 16：面向对象基础：class 与 __init__

类把"数据"和"操作数据的方法"打包成对象。
`__init__` 是构造方法（创建对象时自动调用），`self` 指代实例自身。

```python
class Dog:
    def __init__(self, name):
        self.name = name          # 实例属性

    def bark(self):
        return f"{self.name}：汪汪！"

d = Dog("旺财")
print(d.name)
print(d.bark())
```

## Day 17：类的方法与属性

- 实例方法第一个参数固定是 `self`
- `@property` 把方法变成"像属性一样访问"（常用于计算字段）
- 类属性写在方法外，所有实例共享

```python
class Circle:
    pi = 3.14159                   # 类属性（所有实例共享）

    def __init__(self, r):
        self.r = r

    @property
    def area(self):                # 像属性一样访问：c.area
        return Circle.pi * self.r ** 2

c = Circle(2)
print(c.area)
```

## Day 18：继承：复用与重写

子类继承父类，自动获得父类能力，并可重写（override）方法。
`super()` 用于调用父类方法。

```python
class Animal:
    def speak(self):
        return "..."

class Cat(Animal):
    def speak(self):
        return "喵"

class Dog(Animal):
    def speak(self):
        return "汪"

for a in [Cat(), Dog()]:
    print(a.speak())
```

## Day 19：生成器与 yield（按需产出，省内存）

含 `yield` 的函数是生成器：每次产出值后暂停，下次继续。适合处理大数据或流式数据，不会一次性占满内存。

生成器表达式 `(x for x in ...)` 类似列表推导式但更省内存。

```python
def count_up(n):
    i = 0
    while i < n:
        yield i
        i += 1

for x in count_up(5):
    print(x)

# 生成器表达式（注意是 () 不是 []）
squares = (x * x for x in range(5))
print(list(squares))
```

## Day 20：装饰器入门：给函数“加功能”

装饰器是"包装函数"的函数，能在不修改原函数代码的前提下添加日志、计时、权限校验等功能。

本质是：`@timer` 等价于 `slow = timer(slow)`。

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        print(f"{func.__name__} 用时 {time.time() - start:.4f}s")
        return result
    return wrapper

@timer
def slow():
    sum(i * i for i in range(100000))

slow()
```

## Day 22：正则表达式 re（提取与校验文本）

`re` 用于按模式匹配/提取文本。最常用：
`re.findall` 找出所有匹配、`re.search` 查找（可分组 `()`）、`re.fullmatch` 完整校验。

常用元字符：`\d` 数字、`\w` 字母数字、`+` 一个或多个、`*` 零个或多个、`?` 可选。

```python
import re

text = "订单号 12345，电话 13800001234"
nums = re.findall(r"\d+", text)         # 提取所有连续数字
print(nums)

m = re.search(r"(\d{5})", text)         # 提取 5 位数字并捕获分组
print(m.group(1))

print(bool(re.fullmatch(r"1\d{10}", "13800001234")))   # 校验手机号
```

## Day 23：collections：更实用的容器

标准库 `collections` 提供三种高频工具：
- `Counter`：快速计数
- `defaultdict`：访问不存在的 key 时自动给默认值（免去 `if key not in d`）
- `namedtuple`：带名字的元组，像小对象一样用 `.x` 访问

```python
from collections import Counter, defaultdict, namedtuple

words = ["a", "b", "a", "c", "a", "b"]
print(Counter(words))                  # 统计频次

d = defaultdict(int)
for w in words:
    d[w] += 1                          # 无需先判断 key 是否存在
print(dict(d))

Point = namedtuple("Point", ["x", "y"])
pt = Point(3, 4)
print(pt.x, pt.y)
```

## Day 24：JSON 处理

`json` 模块用于序列化/反序列化（和 Web API、配置文件打交道必用）。

- `json.dumps(obj)` 对象 → 字符串；`json.loads(str)` 字符串 → 对象
- `json.dump/load` 直接读写文件
- **中文记得加 `ensure_ascii=False`**，否则中文会被转成 `\uXXXX`

```python
import json

data = {"name": "小明", "age": 18, "skills": ["Python", "SQL"]}
s = json.dumps(data, ensure_ascii=False, indent=2)    # 中文不转义
print(s)

back = json.loads(s)
print(back["skills"])

with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
```

## Day 28：Pythonic 习惯用法（让代码更地道）

一些让代码更简洁、更"Python 风格"的写法：
- `in` 判断成员
- 解包：`a, b, *rest = [...]`
- 一行交换：`a, b = b, a`
- 海象运算符 `:=`（3.8+）：在表达式里顺手赋值

```python
nums = [1, 2, 3]
if 2 in nums:                      # 推荐，别写 index 再比较
    print("存在")

a, b, *rest = [1, 2, 3, 4, 5]     # 解包，rest 收剩余
print(a, b, rest)

x, y = 1, 2
x, y = y, x                       # 交换变量，无需临时变量
print(x, y)

data = [10, 20, 30]
if (n := len(data)) > 2:          # 海象运算符：赋值并判断
    print(f"长度 {n} 大于 2")
```

## Day 29：实战（一）：文件批量扫描器

综合运用前面知识：用 `pathlib` 递归遍历目录、统计文件、按后缀分类。

```python
from pathlib import Path
from collections import Counter

def scan_dir(folder):
    root = Path(folder)
    if not root.exists():
        return []
    return [p for p in root.rglob("*") if p.is_file()]   # 递归找所有文件

files = scan_dir(".")             # 扫描当前目录
print(f"共找到 {len(files)} 个文件")
ext_counter = Counter(p.suffix for p in files)
print("各后缀数量：", dict(ext_counter))
for f in files[:5]:
    print("  ", f.name)
```

## 练习：综合小项目 —— 待办清单（To-Do List）

**目标**：用一个列表管理任务，综合运用 列表 / 字典 / 函数 / 循环 / 字符串格式化 / 文件读写。

**要求**：
1. 用列表 `tasks` 存放任务，每个任务是一个字典 `{"done": False, "title": "..."}`。
2. 实现 `add_task(title)`：添加任务。
3. 实现 `complete_task(index)`：把第 `index` 个任务标记为完成。
4. 实现 `list_tasks()`：打印所有任务，完成的打 ✅、未完成的打 ⬜，并带序号。
5. 实现 `remove_task(index)`：删除指定任务。
6. 写一段演示：添加 3 个任务 → 完成其中 1 个 → 列出 → 删除 1 个。

**进阶（可选）**：用 Day 13 的 `open/with` 把清单保存到 `todos.txt`，每行一个任务。

先自己写一遍，再对照下面的参考实现 👇


```python
# ===== 参考实现（先自己写，再对照） =====
tasks = []

def add_task(title):
    tasks.append({"done": False, "title": title})

def complete_task(index):
    if 0 <= index < len(tasks):
        tasks[index]["done"] = True

def remove_task(index):
    if 0 <= index < len(tasks):
        tasks.pop(index)

def list_tasks():
    if not tasks:
        print("（清单为空）")
        return
    for i, t in enumerate(tasks):
        mark = "✅" if t["done"] else "⬜"
        print(f"{i}. {mark} {t['title']}")

# 演示
add_task("学 Python 切片")
add_task("写一个函数")
add_task("读 30 天笔记")
complete_task(0)
print("--- 当前清单 ---")
list_tasks()
remove_task(2)
print("--- 删除后 ---")
list_tasks()

# 进阶：保存到文件
with open("todos.txt", "w", encoding="utf-8") as f:
    for t in tasks:
        f.write(("✅ " if t["done"] else "⬜ ") + t["title"] + "\n")
print("\n已保存到 todos.txt")
```
