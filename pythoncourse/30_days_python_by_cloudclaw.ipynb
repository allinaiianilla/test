## Day 2：字符串操作

**今日目标**：掌握字符串格式化、切片、常用方法。

### 核心概念
- f-string 是最推荐的格式化方式（Python 3.6+）
- 切片 `[start:end:step]` 是 Python 标志性操作
- `split/join/strip/replace` 是四大字符串方法

```python
# === Day 2 实操 ===

# 1. f-string — 最常用格式化
name, score = "Alice", 98.5
print(f"{name} 考了 {score} 分")

# 2. 切片 — Python 必杀技
s = "Hello Python"
print(s[0:5])    # Hello
print(s[:5])     # Hello（从头开始可省略0）
print(s[6:])     # Python（到最后可省略末尾）
print(s[::-1])   # nohtyP olleH（反转字符串！）

# 3. 四大字符串方法
raw = "  apple,banana ,cherry  "
print(raw.strip())              # 去首尾空格
print(raw.replace("banana", "orange"))  # 替换
print(raw.strip().split(","))  # 分割 → ['apple', 'banana ', 'cherry']

fruits = ["A", "B", "C"]
print("|".join(fruits))        # 拼接 → A|B|C

# 4. 判断方法
print("123".isdigit())  # True
print("abc".isalpha())  # True
print("Hello".startswith("He"))  # True
```

## Day 3：列表 — 最万能的数据容器

**今日目标**：列表的增删改查、遍历、排序。

### 核心概念
- 列表是**有序、可变**的容器
- 增：`append/extend/insert` | 删：`remove/pop`
- `enumerate()` 同时获取索引和值

```python
# === Day 3 实操 ===

# 1. 增删改查
items = [1, 2, 3]
items.append(4)      # [1,2,3,4] — 尾部加
items.insert(0, 0)   # [0,1,2,3,4] — 指定位置加
items.extend([5,6])  # [0,1,2,3,4,5,6] — 合并列表

items.remove(0)      # [1,2,3,4,5,6] — 按值删
last = items.pop()   # last=6, items=[1,2,3,4,5] — 弹尾
items[2] = 99        # [1,2,99,4,5] — 改值

print(items)
print(items.index(99))  # 2 — 查找位置
print(5 in items)       # True — 是否存在

# 2. 遍历（工作中最最常用的模式）
for i, val in enumerate(items):
    print(f"索引{i} → {val}")

# 3. 切片（列表同样适用）
print(items[1:3])   # [2, 99]
print(items[::-1])  # [5, 4, 99, 2, 1] — 反转

# 4. 排序
nums = [3, 1, 4, 1, 5, 9]
print(sorted(nums))           # [1,1,3,4,5,9] 不改变原列表
print(sorted(nums, reverse=True))  # 降序
nums.sort()                   # 原地排序
print(nums)
```

## Day 4：元组、集合、字典

**今日目标**：三大数据结构的区分与适用场景。

### 核心概念
| 类型 | 特点 | 场景 |
|------|------|------|
| tuple | 不可变 | 常量、函数多返回、字典键 |
| set | 无序、去重 | 去重、交并差运算 |
| dict | 键值对 | 配置、计数、缓存、JSON |

```python
# === Day 4 实操 ===

# 1. 元组 — 不可变的列表
point = (3, 4)
x, y = point   # 解包，超常用！
print(f"x={x}, y={y}")

# 2. 集合 — 去重神器
nums = [1, 2, 2, 3, 3, 3]
unique = list(set(nums))
print(unique)  # [1, 2, 3]

A = {1, 2, 3}
B = {3, 4, 5}
print(A & B)   # {3} — 交集
print(A | B)   # {1,2,3,4,5} — 并集
print(A - B)   # {1, 2} — 差集

# 3. 字典 — 80% 的 Python 代码都在用
user = {
    "name": "张三",
    "age": 28,
    "skills": ["Python", "SQL"]
}

# 安全取值（避免 KeyError）
print(user.get("email", "未设置"))  # 未设置

# 遍历字典
for key, value in user.items():
    print(f"{key} → {value}")

# 合并字典
extra = {"city": "北京", "age": 30}
merged = {**user, **extra}  # 后者覆盖前者
print(merged)

# 计数模式（面试常考）
words = ["a", "b", "a", "c", "b", "a"]
count = {}
for w in words:
    count[w] = count.get(w, 0) + 1
print(count)  # {'a': 3, 'b': 2, 'c': 1}
```

## Day 6：for / while 循环

**今日目标**：掌握两种循环的场景区分、break/continue。

### 核心概念
- `for` 遍历可迭代对象（90% 的场景用它）
- `while` 条件循环（不确定次数的场景）
- `range()` 生成数字序列
- `break` 跳出循环，`continue` 跳过本次

```python
# === Day 6 实操 ===

# 1. for + range（遍历数字）
for i in range(5):        # 0,1,2,3,4
    print(i, end=" ")
print()

for i in range(2, 6):     # 2,3,4,5
    print(i, end=" ")
print()

# 2. break — 找到就停（效率优化关键）
data = [1, 3, 5, 7, -1, 9]
for d in data:
    if d < 0:
        print(f"发现负数: {d}")
        break

# 3. continue — 跳过不符合条件的
for n in range(10):
    if n % 2 == 0:
        continue  # 跳过偶数
    print(n, end=" ")  # 1 3 5 7 9
print()

# 4. while — 读文件/轮询等场景
count = 0
while count < 3:
    print(f"第{count+1}次尝试")
    count += 1

# 5. for + else（Python 特有）— 循环正常结束才执行
for x in [1, 2, 3]:
    if x > 10:
        break
else:
    print("没找到大于10的数（for循环正常结束）")
```

## Day 7：列表/字典推导式 — Python 的效率之魂

**今日目标**：一行代码替代多行循环，Python 程序员必备技能。

### 核心概念
- 列表推导式：`[表达式 for 变量 in 可迭代对象 if 条件]`
- 字典推导式：`{key: value for ...}`
- 集合推导式：`{表达式 for ...}`

```python
# === Day 7 实操 ===

# 1. 列表推导式 — 三行变一行
# 传统写法
sq1 = []
for i in range(10):
    if i % 2 == 0:
        sq1.append(i ** 2)

# 推导式写法
sq2 = [i**2 for i in range(10) if i % 2 == 0]
print(sq2)  # [0, 4, 16, 36, 64]

# 2. 字典推导式 — 日常超高频
names = ["Alice", "Bob", "Charlie"]
name_len = {name: len(name) for name in names}
print(name_len)  # {'Alice': 5, 'Bob': 3, 'Charlie': 7}

# 翻转字典
d = {"a": 1, "b": 2, "c": 3}
flipped = {v: k for k, v in d.items()}
print(flipped)  # {1: 'a', 2: 'b', 3: 'c'}

# 3. 带条件过滤
nums = [1, 2, 3, 4, 5, 6]
evens = [n for n in nums if n % 2 == 0]
print(evens)

# 4. 嵌套推导式（如展平二维列表）
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [x for row in matrix for x in row]
print(flat)  # [1, 2, 3, 4, 5, 6]

# 5. 生成器表达式 — 省内存
# 把方括号换成圆括号即可
big_sum = sum(i**2 for i in range(1000000))  # 不会生成 100 万的中间列表
print(f"平方和: {big_sum}")
```

## Day 8：zip / enumerate / sorted — 最常用的三个内置函数

**今日目标**：三个函数搞定日常数据处理。

```python
# === Day 8 实操 ===

# 1. enumerate — 遍历时带索引
fruits = ["苹果", "香蕉", "橘子"]
for i, fruit in enumerate(fruits):
    print(f"#{i+1} {fruit}")

# 2. zip — 同时遍历多个列表
names = ["张三", "李四", "王五"]
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f"{name}: {score}分")

# 拆 zip（解压）
paired = list(zip(names, scores))
unzip_names, unzip_scores = zip(*paired)
print(unzip_names)   # ('张三', '李四', '王五')
print(unzip_scores)  # (85, 92, 78)

# 3. sorted 排序
students = [
    {"name": "张三", "score": 85},
    {"name": "李四", "score": 92},
    {"name": "王五", "score": 78},
]
# 按 score 降序
ranked = sorted(students, key=lambda s: s["score"], reverse=True)
for s in ranked:
    print(f"{s['name']}: {s['score']}")

# 4. any / all
print(any([False, False, True]))  # True（有真即可）
print(all([True, True, False]))   # False（全真才行）
print(all(80 <= s["score"] <= 100 for s in students))  # 所有人成绩>=80?
```

## Day 10：lambda 与高阶函数

**今日目标**：理解 lambda、map/filter/reduce、闭包基础。

### 核心概念
- lambda 就是**一次性匿名函数**
- 排序 `key=`、`map/filter`、pandas `apply` 是最主要的 lambda 场景
- 闭包：函数里套函数，内层记住外层变量

```python
# === Day 10 实操 ===

# 1. lambda — 短小精悍
add = lambda x, y: x + y
print(add(3, 5))  # 8

# 排序花样（lambda 最常见使用场景）
pairs = [(1, "b"), (2, "a"), (3, "c")]
print(sorted(pairs, key=lambda p: p[1]))          # 按第二个元素排
print(sorted(pairs, key=lambda p: p[0], reverse=True))  # 降序

# 2. map — 对每个元素做同一操作
nums = [1, 2, 3, 4, 5]
print(list(map(lambda x: x**2, nums)))  # [1, 4, 9, 16, 25]
# 但推导式更 Pythonic: [x**2 for x in nums]

# 3. filter — 过滤
print(list(filter(lambda x: x > 3, nums)))  # [4, 5]
# 推导式更 Pythonic: [x for x in nums if x > 3]

# 4. 闭包 — 函数工厂
def make_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(5))  # 10
print(triple(5))  # 15
```

## Day 11：函数进阶 — 装饰器、类型注解

**今日目标**：理解装饰器本质、会写简单装饰器、了解类型注解。

### 核心概念
- 装饰器 = 给函数包一层，不改变原函数，加额外功能
- `@timer` 就是 `func = timer(func)` 的语法糖
- 类型注解让代码更易读，IDE 更智能

```python
# === Day 11 实操 ===

import time
from functools import wraps
from typing import Callable, List, Dict

# 1. 计时装饰器 — 最实用的装饰器
def timer(func: Callable) -> Callable:
    @wraps(func)  # 保留原函数元信息
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        elapsed = time.time() - start
        print(f"[{func.__name__}] 耗时: {elapsed:.4f}秒")
        return result
    return wrapper

@timer
def slow_sum(n):
    time.sleep(0.1)
    return sum(range(n))

print(slow_sum(1000000))

# 2. 重试装饰器
def retry(max_attempts=3):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    if attempt == max_attempts:
                        raise
                    print(f"  重试 {attempt}/{max_attempts}: {e}")
                    time.sleep(0.5)
        return wrapper
    return decorator

# 3. 类型注解 — 让代码自带文档
def calculate_average(scores: List[float]) -> float:
    """计算平均分"""
    return sum(scores) / len(scores) if scores else 0.0

result: float = calculate_average([85.5, 92.0, 78.5])
print(f"平均分: {result:.1f}")
```

## Day 12：异常处理

**今日目标**：写出健壮的代码，不乱崩。

### 核心概念
- `try/except/finally` 是必备技能
- **捕获具体异常**，不要裸 `except:`
- `finally` 做清理，`else` 无异常时执行
- 自定义异常让错误更清晰

```python
# === Day 12 实操 ===

# 1. 基本异常处理
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print(f"错误: 除数不能为零 (a={a}, b={b})")
        return None
    except TypeError as e:
        print(f"类型错误: {e}")
        return None
    finally:
        # 清理操作（如关闭文件、释放锁等）

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # None

# 2. 实际场景：解析用户输入
def parse_age(user_input):
    try:
        age = int(user_input)
        if age < 0 or age > 150:
            raise ValueError(f"年龄不合理: {age}")
        return age
    except ValueError as e:
        print(f"输入错误: {e}")
        return None

print(parse_age("25"))      # 25
print(parse_age("abc"))     # None
print(parse_age("200"))     # None

# 3. 自定义异常
class ConfigError(Exception):
    """配置相关错误"""
    pass

def load_config(key):
    config = {"host": "localhost"}
    if key not in config:
        raise ConfigError(f"缺少配置项: {key}")
    return config[key]

try:
    load_config("port")
except ConfigError as e:
    print(f"配置错误: {e}")

# 4. 上下文管理器模式（with 语句）
# with 自动调用 __enter__ / __exit__，确保资源释放
from contextlib import contextmanager

@contextmanager
def timer_context(name="操作"):
    start = time.time()
    try:
        yield  # 这里执行 with 块里的代码
    finally:
        print(f"[{name}] 耗时: {time.time()-start:.3f}秒")

with timer_context("数据处理"):
    total = sum(range(1000000))
    print(f"结果: {total}")
```

## Day 14：JSON 处理 — 数据交换之王

```python
# === Day 14 实操 ===

import json
from datetime import datetime, date

# 1. 基本转换：dict ↔ JSON
data = {
    "name": "张三",
    "age": 28,
    "skills": ["Python", "SQL", "Docker"],
    "active": True,
    "score": None
}

# Python → JSON 字符串
json_str = json.dumps(data, ensure_ascii=False, indent=2)
print("JSON:")
print(json_str)

# JSON 字符串 → Python
parsed = json.loads(json_str)
print(f"\n解析: {parsed['name']}")

# 2. 读写 JSON 文件
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

with open("data.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)
    print(f"从文件读取: {loaded['skills']}")

# 3. 处理 datetime（JSON 不认识）
class DateTimeEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, (datetime, date)):
            return obj.isoformat()
        return super().default(obj)

log = {"event": "login", "time": datetime.now()}
print(json.dumps(log, cls=DateTimeEncoder, ensure_ascii=False))

# 清理
os.remove("data.json")
```

## 🚀 Day 14 今日成果：API 数据缓存器

**学完今天的内容，你就能做这个 👇**

调用 API 并缓存为 JSON，避免重复请求

```python
# 🚀 Day 14 成果：API 数据缓存器
import json, time
from pathlib import Path

CACHE_FILE = Path("api_cache.json")
CACHE_TTL = 300  # 缓存 5 分钟

def fetch_with_cache(url):
    now = time.time()
    if CACHE_FILE.exists():
        cache = json.loads(CACHE_FILE.read_text())
        if now - cache["ts"] < CACHE_TTL:
            print(f"📦 命中缓存! ({(now-cache['ts']):.0f}秒前)")
            return cache["data"]
    print(f"🌐 请求 {url}...")
    data = {"url": url, "result": "模拟数据", "time": time.ctime()}
    CACHE_FILE.write_text(
        json.dumps({"ts": now, "data": data}, indent=2, ensure_ascii=False),
        encoding="utf-8")
    print("✅ 已缓存")
    return data

print(fetch_with_cache("https://api.github.com"))
print(f"\n缓存文件: {CACHE_FILE.resolve()}")
if CACHE_FILE.exists(): CACHE_FILE.unlink()
```

```python
# 🚀 Day 18 成果：文件整理器
import shutil
from pathlib import Path

def organize_files(directory):
    d = Path(directory)
    if not d.exists(): return
    moved = 0
    for file in d.iterdir():
        if file.is_file():
            ext = file.suffix.lower().lstrip(".") or "other"
            subdir = d / ext
            subdir.mkdir(exist_ok=True)
            shutil.move(str(file), str(subdir / file.name))
            print(f"  📁 {file.name} → {ext}/")
            moved += 1
    print(f"\n✅ 整理了 {moved} 个文件")

# 创建测试文件
Path("demo").mkdir(exist_ok=True)
for f in ["a.txt", "b.jpg", "c.txt", "d.pdf", "e.py"]:
    Path(f"demo/{f}").write_text("")

print("📂 整理前:")
for f in sorted(Path("demo").iterdir()):
    print(f"  {f.name}")

organize_files("demo")

print("\n📂 整理后:")
for d in sorted(Path("demo").iterdir()):
    print(f"  📁 {d.name}/")

shutil.rmtree("demo", ignore_errors=True)
```

## Day 19：正则表达式

**今日目标**：掌握最常用的 10 个正则模式，文本处理利器。

### 核心概念
- 正则就是**模式匹配语言**
- `re.search/findall/sub` 是三巨头
- **不用背**，忘了查表就行，但理解基本规则

```python
# === Day 19 实操 ===

import re

print("=== 10 个最常用的正则模式 ===\n")

# 1. 提取数字
text = "订单号: ORD-2024-00123, 金额: ¥199.99"
nums = re.findall(r'\d+\.?\d*', text)
print(f"1. 提取数字: {nums}")

# 2. 提取邮箱
text2 = "联系: alice@example.com 和 bob@test.cn"
emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text2)
print(f"2. 提取邮箱: {emails}")

# 3. 提取手机号
text3 = "手机: 13800138000, 座机: 010-12345678"
phones = re.findall(r'1[3-9]\d{9}', text3)
print(f"3. 提取手机号: {phones}")

# 4. 提取中文
text4 = "Python 是一种编程语言，release 于 1991 年"
chinese = re.findall(r'[\u4e00-\u9fff]+', text4)
print(f"4. 提取中文: {chinese}")

# 5. 替换 — sub
text5 = "密码是 123456，千万别泄露"
masked = re.sub(r'\d+', '****', text5)
print(f"5. 脱敏替换: {masked}")

# 6. 分割
text6 = "张三,李四;王五|赵六"
parts = re.split(r'[,;|]', text6)
print(f"6. 多符号分割: {parts}")

# 7. search — 找到第一个匹配
match = re.search(r'\d+', "价格是 199 元")
if match:
    print(f"7. search: {match.group()} 在位置 {match.start()}")

# 8. 命名分组
log = "2024-01-15 10:30:45 ERROR 连接超时"
m = re.search(
    r'(?P<date>\d{4}-\d{2}-\d{2}) (?P<time>\d{2}:\d{2}:\d{2}) (?P<level>\w+) (?P<msg>.+)',
    log
)
if m:
    print(f"8. 命名分组: {m.groupdict()}")

# 9. 贪婪 vs 非贪婪
html = "<title>我的网页</title><body>内容</body>"
print(f"9a. 贪婪: {re.findall(r'<.*>', html)}")
print(f"9b. 非贪婪: {re.findall(r'<.*?>', html)}")

# 10. 预编译（高频匹配时提升性能）
phone_pattern = re.compile(r'1[3-9]\d{9}')
print(f"10. 预编译匹配: {phone_pattern.findall('13800138000, 13900139000')}")

print("\n常用速查: \\d=数字 \\w=字母数字 \\s=空白 .*=任意次数 +=1次+ ?=可选 [] =字符组 ()=分组")
```

算两个日期差、距离某天还有多久

```python
# 🚀 Day 20 成果：倒计时计算器
from datetime import datetime, date

def days_between(d1_str, d2_str):
    d1 = datetime.strptime(d1_str, "%Y-%m-%d").date()
    d2 = datetime.strptime(d2_str, "%Y-%m-%d").date()
    return abs((d2 - d1).days)

def countdown(target_str):
    target = datetime.strptime(target_str, "%Y-%m-%d").date()
    today = date.today()
    return (target - today).days

print("📅 日期计算器")
print(f"  元旦到今天: {days_between('2024-01-01', '2024-03-15')} 天")
print(f"  出生到现在: {days_between('1995-06-15', '2024-03-15')} 天")
print(f"  到2025元旦还有: {countdown('2025-01-01')} 天")

age_days = days_between("1995-06-15", "2024-03-15")
print(f"\n🎂 如果1995-06-15出生，已活 {age_days:,} 天")
```

## Day 21：面向对象基础 — 类与实例

**今日目标**：类的基本语法、`__init__`、实例方法。

### 核心概念
- 类 = 蓝图，实例 = 按蓝图造出来的对象
- `self` 代表实例本身，是 Python 的约定（不是关键字）
- 属性 + 方法 = 数据 + 行为

```python
# === Day 21 实操 ===

class User:
    """用户类 — 最基础的 OOP 例子"""

    # 类属性（所有实例共享）
    platform = "Web"

    def __init__(self, name, age, email=None):
        """构造方法 — 创建实例时自动调用"""
        self.name = name      # 实例属性
        self.age = age
        self.email = email or f"{name}@example.com"
        self._login_count = 0  # 下划线开头 = 约定私有

    def login(self):
        """实例方法"""
        self._login_count += 1
        return f"{self.name} 第{self._login_count}次登录"

    def __str__(self):
        """打印对象的友好表示"""
        return f"User({self.name}, {self.age}岁)"

    def __repr__(self):
        """调试表示"""
        return f"User(name='{self.name}', age={self.age})"

# 使用
u1 = User("张三", 28)
u2 = User("李四", 32, "lisi@qq.com")

print(u1)                    # User(张三, 28岁)
print(repr(u1))             # User(name='张三', age=28)
print(u1.login())           # 张三 第1次登录
print(u1.login())           # 张三 第2次登录
print(f"邮箱: {u1.email}")  # 张三@example.com
print(f"邮箱: {u2.email}")  # lisi@qq.com
print(f"平台: {User.platform}")  # 类属性
```

## 🚀 Day 21 今日成果：银行账户系统

**学完今天的内容，你就能做这个 👇**

用类封装存钱取钱逻辑

```python
# 🚀 Day 21 成果：银行账户系统
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.history = []

    def deposit(self, amount):
        if amount <= 0: return "金额必须>0"
        self.balance += amount
        self.history.append(f"+{amount}")
        return f"存入 ¥{amount:,}"

    def withdraw(self, amount):
        if amount <= 0: return "金额必须>0"
        if amount > self.balance:
            return f"余额不足！仅 ¥{self.balance:,}"
        self.balance -= amount
        self.history.append(f"-{amount}")
        return f"取出 ¥{amount:,}"

    def __str__(self):
        return f"{self.owner} | 余额 ¥{self.balance:,} | {len(self.history)}笔"

acc = BankAccount("张三", 10000)
print(acc)
print(acc.deposit(5000))
print(acc.withdraw(3000))
print(acc.withdraw(20000))
print(f"\n交易记录: {acc.history}")
print(acc)
```

## Day 22：继承与多态

**今日目标**：父子类关系、方法重写、`super()`。

```python
# === Day 22 实操 ===

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} 发出声音"

    def move(self):
        return "移动"


class Dog(Animal):
    def speak(self):          # 方法重写
        return f"{self.name}: 汪汪!"

    def fetch(self):          # 子类特有方法
        return "叼回骨头"


class Cat(Animal):
    def speak(self):
        return f"{self.name}: 喵喵~"


# 多态 — 同一接口，不同行为
animals = [Dog("旺财"), Cat("小花"), Dog("阿黄")]
for a in animals:
    print(a.speak())

# super() — 调父类方法
class RobotDog(Dog):
    def __init__(self, name, battery):
        super().__init__(name)  # 调父类 __init__
        self.battery = battery

    def speak(self):
        parent_msg = super().speak()  # 调父类 speak
        return f"🤖 {parent_msg} (电量:{self.battery}%)"


robot = RobotDog("铁蛋", 80)
print(robot.speak())

# isinstance 检查继承关系
print(f"robot 是 Dog? {isinstance(robot, Dog)}")       # True
print(f"robot 是 Animal? {isinstance(robot, Animal)}") # True (间接继承)
```

用继承实现不同类型商品的折扣

```python
# 🚀 Day 22 成果：商城折扣系统
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def final_price(self): return self.price
    def tag(self): return f"{self.name} ¥{self.final_price():.0f}"

class DiscountProduct(Product):
    def __init__(self, name, price, discount=0.8):
        super().__init__(name, price)
        self.discount = discount

    def final_price(self): return self.price * self.discount
    def tag(self): return f"{self.name} ¥{self.final_price():.0f} (原¥{self.price})"

class VipProduct(DiscountProduct):
    def final_price(self): return self.price * self.discount * 0.9

items = [
    Product("Python书", 99),
    DiscountProduct("键盘", 399, 0.7),
    VipProduct("显示器", 1999, 0.8),
]

print("🛒 购物清单:")
total = sum(i.final_price() for i in items)
for item in items: print(f"  {item.tag()}")
print(f"\n💰 合计: ¥{total:,.0f} (原价 ¥{sum(i.price for i in items):,.0f})")
```

## Day 23：魔术方法与属性

**今日目标**：`__len__/__getitem__/__iter__`、`@property`。

### 核心概念
- 魔术方法让你的类像内置类型一样工作
- `@property` 把方法变成属性访问

```python
# === Day 23 实操 ===

class Playlist:
    def __init__(self, name, songs=None):
        self.name = name
        self._songs = songs or []

    # 让对象支持 len()
    def __len__(self):
        return len(self._songs)

    # 让对象支持 [] 索引
    def __getitem__(self, index):
        return self._songs[index]

    # 让对象可迭代
    def __iter__(self):
        return iter(self._songs)

    # 让对象支持 in 操作
    def __contains__(self, item):
        return item in self._songs

    # 让对象支持 +
    def __add__(self, other):
        return Playlist(
            f"{self.name}+{other.name}",
            self._songs + other._songs
        )


pl = Playlist("最爱", ["晴天", "七里香", "夜曲"])
print(f"长度: {len(pl)}")
print(f"第一首: {pl[0]}")
print("\"夜曲\" 在列表里?", "夜曲" in pl)

# 合并
pl2 = Playlist("新歌", ["告白气球"])
merged = pl + pl2
print(f"合并: {list(merged)}")

# === @property — getter/setter 更优雅 ===

class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price

    @property
    def price(self):
        """像属性一样访问"""
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("价格不能为负")
        self._price = value

    @property
    def price_with_tax(self):
        """计算属性"""
        return round(self._price * 1.13, 2)


p = Product("《Python入门》", 59.0)
print(f"价格: ¥{p.price}, 含税: ¥{p.price_with_tax}")
p.price = 69.0  # 通过 setter 赋值
print(f"新价格: ¥{p.price}")
```

用@property优雅实现多单位换算

```python
# 🚀 Day 23 成果：温度转换器
class Temperature:
    def __init__(self, celsius=0):
        self.celsius = celsius

    @property
    def fahrenheit(self):
        return self.celsius * 9/5 + 32

    @fahrenheit.setter
    def fahrenheit(self, value):
        self.celsius = (value - 32) * 5/9

    @property
    def kelvin(self):
        return self.celsius + 273.15

    def __str__(self):
        return f"{self.celsius:.1f}°C = {self.fahrenheit:.1f}°F = {self.kelvin:.2f}K"

t = Temperature(25)
print(f"初始: {t}")

t.fahrenheit = 100
print(f"100°F → {t}")

t.celsius = 0
print(f"冰点 → {t}")

t.fahrenheit = 212
print(f"沸点 → {t}")
```

## Day 24：collections — 进阶数据结构

**今日目标**：Counter、defaultdict、OrderedDict、namedtuple。

```python
# === Day 24 实操 ===

from collections import Counter, defaultdict, OrderedDict, namedtuple

# 1. Counter — 计数器（统计频率神器）
words = list("abracadabra")
c = Counter(words)
print(f"字母频率: {c}")
print(f"最常见2个: {c.most_common(2)}")

# 2. defaultdict — 有默认值的字典
# 不用再写: d[key] = d.get(key, []) + ...
groups = defaultdict(list)
data = [("A", 1), ("B", 2), ("A", 3), ("C", 4), ("B", 5)]
for key, val in data:
    groups[key].append(val)
print(f"分组: {dict(groups)}")

# 默认数量
counts = defaultdict(int)
for w in ["a", "b", "a", "c", "b", "a"]:
    counts[w] += 1  # 不用先判断 key 是否存在！
print(f"计数: {dict(counts)}")

# 3. namedtuple — 像字典一样可读，像元组一样轻量
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(f"x={p.x}, y={p.y}")
print(f"距离原点: {(p.x**2 + p.y**2)**0.5}")

# 4. OrderedDict — 保持插入顺序（Python 3.7+ 普通 dict 也保序了）
od = OrderedDict()
od["first"] = 1
od["second"] = 2
print(f"有序字典: {list(od.keys())}")
```

## 🚀 Day 25 今日成果：批量文件重命名器

把目录下的文件统一加前缀/序号

```python
# 🚀 Day 25 成果：批量文件重命名器
from pathlib import Path

def batch_rename(directory, prefix="file", start=1):
    d = Path(directory)
    files = sorted([f for f in d.iterdir() if f.is_file()])
    for i, f in enumerate(files, start):
        new_name = f"{prefix}_{i:03d}{f.suffix}"
        f.rename(d / new_name)
        print(f"  {f.name} → {new_name}")
    print(f"\n✅ 重命名了 {len(files)} 个文件")

# 创建测试文件
d = Path("test_rename"); d.mkdir(exist_ok=True)
for name in ["photo1.jpg", "photo2.jpg", "video1.mp4"]:
    (d / name).write_text("")

print("重命名前:")
for f in sorted(d.iterdir()): print(f"  {f.name}")

print("\n重命名中:")
batch_rename("test_rename", "旅行")

import shutil; shutil.rmtree("test_rename", ignore_errors=True)
```

输入URL爬取标题、描述和链接数

```python
# 🚀 Day 26 成果：网页信息抓取器
try:
    import requests
    from bs4 import BeautifulSoup

    def get_page_info(url):
        try:
            headers = {"User-Agent": "Mozilla/5.0"}
            r = requests.get(url, headers=headers, timeout=10)
            r.raise_for_status()
            soup = BeautifulSoup(r.text, "html.parser")
            title = soup.title.string.strip() if soup.title else "无标题"
            desc_tag = soup.find("meta", attrs={"name": "description"})
            desc = desc_tag["content"][:100] if desc_tag else "无描述"
            links = len(soup.find_all("a"))
            return {"title": title, "description": desc, "links": links}
        except Exception as e:
            return {"error": str(e)}

    info = get_page_info("https://httpbin.org")
    for k, v in info.items():
        print(f"  {k}: {v}")
except ImportError:
    print("请先: pip install requests beautifulsoup4")
```

## 🏗️ 开源项目实战：dataclasses（标准库）
### 匹配 Day 21：面向对象 / Day 30：综合实战

**源码片段（dataclass 实战模式）：**

```python
# Python 标准库 dataclasses 的核心用法from dataclasses import dataclass, field, asdictfrom typing import ClassVar@dataclass(order=True, frozen=False)class StockPosition:    """股票持仓数据类"""    symbol: str    shares: int = field(default=0, metadata={"unit": "股"})    avg_cost: float = 0.0    # 类变量（不参与 __init__）    MARKET: ClassVar[str] = "A股"    # 计算属性（每次计算）    @property    def market_value(self) -> float:        return self.shares * self.avg_cost    def pnl(self, current_price: float) -> float:        """浮动盈亏"""        return self.shares * (current_price - self.avg_cost)# 使用pos = StockPosition("贵州茅台", shares=100, avg_cost=1800.0)print(pos.market_value)         # 180000.0print(pos.pnl(1850.0))          # 5000.0print(asdict(pos))              # {'symbol': '贵州茅台', 'shares': 100, 'avg_cost': 1800.0}
```

**讲解：** `dataclasses` 是 Day 21 OOP 和 Day 30 综合实战的完美结合。`@dataclass(order=True)` 自动生成 `__init__`、`__repr__`、`__eq__`、`__lt__` 等方法（Day 21），`field(default=0)` 定义默认值（Day 9 参数默认值），`@property` 定义计算属性（Day 22），`ClassVar` 区分类变量和实例变量（Day 21 类属性 vs 实例属性）。这是 Python 3.7+ 数据建模的首选工具，对比手写 `__init__` 可以减少大量样板代码。

## 🏗️ 开源项目实战：django (⭐82K)
### 匹配 Day 4：字典/查询集 / Day 22：类继承

**源码片段（Django ORM QuerySet 与中间件模式）：**

```python
# Django ORM 的链式操作模式（简化自 django/db/models/query.py 核心逻辑）# QuerySet 是不可变的——每次操作返回新 QuerySet# 链式操作——每一步都返回新的 QuerySetactive_users = (    User.objects    .filter(is_active=True)           # WHERE is_active = True    .exclude(role="banned")           # AND role != 'banned'    .select_related("profile")        # JOIN profile    .prefetch_related("posts")        # 预加载 posts    .order_by("-date_joined")         # ORDER BY date_joined DESC    .only("id", "username", "email")  # 只取需要的字段)# Django 中间件模式class RequestTimingMiddleware:    """记录每个请求的处理时间"""    def __init__(self, get_response):        self.get_response = get_response    def __call__(self, request):        import time        start = time.time()        response = self.get_response(request)        duration = time.time() - start        response["X-Request-Duration-ms"] = str(int(duration * 1000))        return response
```

**讲解：** Django ORM 的链式调用是 Day 4（字典/list操作）和 Day 5（条件过滤）的工业级应用。每个方法返回新 QuerySet（不可变对象模式），允许无限链式组合——这是函数式编程（Day 8 的惰性求值）思想在 OOP 中的体现。`prefetch_related`/`select_related` 解决 N+1 查询问题。中间件 `__call__` 方法（Day 21）让类的实例可调用，在请求前/后插入逻辑——这是 AOP（面向切面编程）在 Python 中的经典实现。

### 匹配 Day 27-28：DataFrame 链式操作

**源码片段（DataFrame 链式操作与 groupby）：**

```python
import pandas as pd# DataFrame 链式操作——每一步都返回新的 DataFrame/Seriesresult = (    pd.read_csv("sales.csv")    .query("amount > 0")                       # 过滤负值    .assign(        month=lambda df: df["date"].dt.month,  # 派生列        revenue=lambda df: df["amount"] * df["price"],    )    .groupby(["region", "month"])              # 分组    .agg(        total_revenue=("revenue", "sum"),        avg_price=("price", "mean"),        orders=("order_id", "nunique"),    )    .query("orders >= 10")                     # 过滤小样本    .sort_values("total_revenue", ascending=False)    .reset_index())# pivot_table 透视表summary = pd.pivot_table(    result,    values="total_revenue",    index="region",    columns="month",    fill_value=0,    margins=True,          # 添加汇总行/列)
```

**讲解：** pandas 的链式操作是 Day 27-28 的核心内容。`.query()` 用字符串表达式过滤（比 `df[df["amount"]>0]` 更可读），`.assign()` 用 lambda 创建派生列（Day 10 lambda），`.groupby().agg()` 的分组聚合是多级字典（Day 4）的应用——`{"total_revenue": ("revenue", "sum")}` 的语法充分体现了 Python 字典的灵活性。整个链式管道（pipe）的思想源自 Unix 哲学：每个步骤做一件事，组合完成复杂任务。
