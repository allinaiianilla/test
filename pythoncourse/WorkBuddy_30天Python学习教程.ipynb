# Day 1: 变量与数据类型

---

## 📖 今日讲解

Python 中最基础的概念就是 **变量** 和 **数据类型**。

你可以把变量想象成一个 **贴了标签的盒子**，盒子里装数据，标签就是变量名。

Python 有四种最常用的内置数据类型：

| 类型 | 英文名 | 例子 | 用途 |
|------|--------|------|------|
| 整数 | int | `10`, `-5`, `0` | 计数、循环 |
| 浮点数 | float | `3.14`, `0.5` | 小数计算 |
| 字符串 | str | `"hello"` | 文本处理 |
| 布尔值 | bool | `True`, `False` | 条件判断 |

Python 是 **动态类型语言**，不需要提前声明类型，赋值时自动推断：

```python
x = 10        # 自动识别为 int
x = "hello"   # 同一个变量可以换成 str
```

用 `type()` 函数可以查看变量的类型。

这在实际开发中非常重要——**类型错误**是最常见的 bug 来源之一。

---

## 💻 可运行代码示例

下面这段代码演示了四种基本数据类型的创建和查看：


````python
# === 四种基本数据类型 ===

# 整数 int —— 用于计数、循环
age = 25
print("年龄:", age)
print("类型:", type(age))

# 浮点数 float —— 用于小数计算
price = 19.99
print("价格:", price)
print("类型:", type(price))

# 字符串 str —— 用于文本处理
name = "张三"
print("姓名:", name)
print("类型:", type(name))

# 布尔值 bool —— 用于条件判断
is_student = True
print("是否学生:", is_student)
print("类型:", type(is_student))
````


类型转换也是高频操作，用 `int()`、`float()`、`str()`、`bool()` 可以转换类型：



````python
# === 类型转换 ===

# 字符串转数字（用户输入时经常用到）
user_input = "100"
num = int(user_input)
print("字符串转整数:", num, "类型:", type(num))

# 数字转字符串（拼接输出时用到）
score = 95
msg = "你的分数是 " + str(score) + " 分"
print(msg)

# 浮点数转整数（取整）
price = 19.99
print("取整:", int(price))

# 数字转布尔（0 为 False，非 0 为 True）
print("0 的布尔值:", bool(0))
print("1 的布尔值:", bool(1))
````

# Day 2: 字符串操作

---

## 📖 今日讲解

字符串是 Python 中 **使用频率最高** 的数据类型。

几乎每个程序都会涉及字符串：处理用户输入、拼接输出消息、解析文本数据、爬虫提取内容……

字符串有三种创建方式：

```python
s1 = '单引号'
s2 = "双引号"
s3 = '''三引号
可以跨行'''
```

字符串最常用的操作有：

- **拼接**：`+` 号拼接
- **格式化**：f-string（最推荐）
- **切片**：`s[起:止]` 截取子串
- **常用方法**：`split()`、`replace()`、`strip()`、`upper()`、`find()`

其中 **f-string** 是 Python 3.6+ 推荐的格式化方式，简洁高效：

```python
name = "张三"
age = 25
msg = f"我叫{name}，今年{age}岁"
```

---

## 💻 可运行代码示例


````python
# === 字符串格式化（f-string 最常用）===

name = "张三"
age = 25
score = 95.5

# f-string 格式化
msg = f"姓名:{name}, 年龄:{age}, 分数:{score}"
print(msg)

# 控制小数位数
print(f"分数保留1位小数: {score:.1f}")

# 补零对齐
num = 5
print(f"编号补零: {num:03d}")
````


字符串切片是高频操作，用 `[起始:结束:步长]` 截取：


````python
# === 字符串切片 ===

s = "Hello, Python!"

# 取前5个字符
print("前5个:", s[0:5])

# 取末尾6个字符（负索引）
print("末尾6个:", s[-6:])

# 反转字符串
print("反转:", s[::-1])

# 每隔2个取一个
print("步长2:", s[::2])
````


字符串的常用方法，每个都频繁用到：


````python
# === 字符串常用方法 ===

s = "  Hello, World  "

# strip() 去除首尾空白（用户输入清洗必用）
print("strip:", s.strip())

# replace() 替换内容
print("replace:", s.replace("World", "Python"))

# split() 分割成列表
csv_line = "张三,25,男,北京"
parts = csv_line.split(",")
print("split:", parts)

# upper/lower 大小写转换
print("upper:", "hello".upper())
print("lower:", "HELLO".lower())

# find() 查找子串位置（找不到返回-1）
print("find:", "Hello".find("ll"))

# startswith / endswith 判断开头结尾
print("endswith:", "test.py".endswith(".py"))
````

# Day 3: 列表 List

---

## 📖 今日讲解

列表是 Python 中 **最常用** 的数据结构，没有之一。

你可以把列表理解为一个 **有序的、可修改的容器**，可以装任意类型的数据：

```python
fruits = ["苹果", "香蕉", "橘子"]
```

列表的核心操作：

| 操作 | 方法 | 说明 |
|------|------|------|
| 增 | `append()` / `insert()` | 末尾添加 / 指定位置插入 |
| 删 | `remove()` / `pop()` | 按值删 / 按位置删 |
| 改 | `lst[i] = x` | 按索引修改 |
| 查 | `lst[i]` / `in` | 按索引取 / 判断是否存在 |
| 排序 | `sort()` / `sorted()` | 原地排序 / 返回新列表 |
| 统计 | `len()` / `count()` / `sum()` | 长度 / 计数 / 求和 |

列表同样支持 **切片** `lst[起:止]`，和字符串切片规则一致。

---

## 💻 可运行代码示例


````python
# === 列表增删改查 ===

fruits = ["苹果", "香蕉", "橘子"]

# 增：末尾添加
fruits.append("葡萄")
print("append后:", fruits)

# 增：指定位置插入
fruits.insert(0, "芒果")
print("insert后:", fruits)

# 改：按索引修改
fruits[0] = "西瓜"
print("修改后:", fruits)

# 删：按值删除
fruits.remove("香蕉")
print("remove后:", fruits)

# 删：按位置删除（pop返回被删的值）
removed = fruits.pop(0)
print(f"pop删除了:{removed}, 剩下:{fruits}")

# 查：判断元素是否存在
print("橘子在列表中:", "橘子" in fruits)
````


列表切片和排序：


````python
# === 列表切片与排序 ===

nums = [3, 1, 4, 1, 5, 9, 2, 6]

# 切片：取前3个
print("前3个:", nums[0:3])

# 切片：取后3个
print("后3个:", nums[-3:])

# 切片：反转
print("反转:", nums[::-1])

# 排序：sorted 返回新列表，不改原列表
new_nums = sorted(nums)
print("sorted新列表:", new_nums)
print("原列表不变:", nums)

# 排序：sort 原地排序
nums.sort()
print("sort原地排序:", nums)

# 降序排序
nums.sort(reverse=True)
print("降序:", nums)

# 统计
print("长度:", len(nums))
print("最大:", max(nums))
print("最小:", min(nums))
print("求和:", sum(nums))
````

# Day 4: 字典 Dict

---

## 📖 今日讲解

字典是 Python 中 **第二常用** 的数据结构，用 **键值对** 存储数据。

字典最大的优势：**查找速度极快**，无论存多少数据，查找都是 O(1)。

```python
person = {"name": "张三", "age": 25}
```

字典 vs 列表：

- 列表：有序、按索引查找、适合存储 **同类数据**
- 字典：键值对、按键查找、适合存储 **结构化数据**

字典核心操作：

| 操作 | 语法 | 说明 |
|------|------|------|
| 增/改 | `d[key] = value` | 键存在则改，不存在则增 |
| 删 | `d.pop(key)` | 删除指定键 |
| 查 | `d[key]` / `d.get(key)` | 取值（get 更安全） |
| 遍历 | `for k, v in d.items()` | 遍历键值对 |
| 判断 | `key in d` | 判断键是否存在 |

> ⚠️ **重要**：用 `d[key]` 取值时，键不存在会报错 KeyError。推荐用 `d.get(key, 默认值)`，键不存在时返回默认值，不报错。

---

## 💻 可运行代码示例


````python
# === 字典增删改查 ===

person = {"name": "张三", "age": 25}

# 增/改
person["city"] = "北京"     # 新增
person["age"] = 26          # 修改
print("增改后:", person)

# 查（两种方式）
print("姓名:", person["name"])           # 直接取，键不存在会报错
print("邮箱:", person.get("email", "未填写"))  # get取，不存在返回默认值

# 删
person.pop("city")
print("删除后:", person)

# 判断键是否存在
print("name键存在:", "name" in person)
````


字典遍历是高频操作，三种遍历方式：


````python
# === 字典遍历 ===

student = {"name": "李四", "age": 20, "score": 90, "city": "上海"}

# 遍历所有键值对（最常用）
print("=== 遍历键值对 ===")
for key, value in student.items():
    print(f"{key}: {value}")

# 遍历所有键
print("=== 遍历键 ===")
for key in student.keys():
    print(key)

# 遍历所有值
print("=== 遍历值 ===")
for value in student.values():
    print(value)
````


字典嵌套——实际开发中非常常见：


````python
# === 字典嵌套 ===

# 学生信息：字典里嵌套列表
student = {
    "name": "王五",
    "scores": [90, 85, 92],
    "info": {"age": 20, "city": "广州"}
}

# 访问嵌套数据
print("姓名:", student["name"])
print("第二科成绩:", student["scores"][1])
print("城市:", student["info"]["city"])

# 多个学生：列表里嵌套字典
students = [
    {"name": "张三", "age": 20},
    {"name": "李四", "age": 21},
    {"name": "王五", "age": 22}
]

# 遍历取出所有名字
for s in students:
    print(s["name"])
````

# Day 5: 元组与集合

---

## 📖 今日讲解

**元组 Tuple** 和 **集合 Set** 使用频率低于列表和字典，但在特定场景下不可替代。

### 元组 Tuple

元组就是 **不可变的列表**，创建后不能修改：

```python
t = (1, 2, 3)
```

元组的用途：
- **函数返回多个值**（最常用）
- **作为字典的键**（列表不行，元组可以）
- **保护数据不被修改**

### 集合 Set

集合是 **无序的、不重复的元素集合**：

```python
s = {1, 2, 3}
```

集合的两大用途：
- **去重**（最常用）
- **集合运算**（交集、并集、差集）

---

## 💻 可运行代码示例


````python
# === 元组 ===

# 创建元组
point = (3, 4)
print("坐标:", point)

# 元组解包（函数返回多值时常用）
x, y = point
print(f"x={x}, y={y}")

# 函数返回多个值
def get_user_info():
    name = "张三"
    age = 25
    return name, age   # 实际返回的是元组

name, age = get_user_info()
print(f"姓名:{name}, 年龄:{age}")
````


集合的去重和运算：


````python
# === 集合 ===

# 去重（最常用场景）
nums = [1, 2, 2, 3, 3, 3, 4]
unique = set(nums)
print("去重后:", unique)
print("转回列表:", list(unique))

# 集合运算
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# 交集（两个集合都有的）
print("交集:", set_a & set_b)

# 并集（合并去重）
print("并集:", set_a | set_b)

# 差集（A有B没有的）
print("差集:", set_a - set_b)

# 添加和删除
set_a.add(5)
print("add后:", set_a)
set_a.discard(1)  # discard不存在不报错，remove不存在会报错
print("discard后:", set_a)
````

# Day 6: 条件判断 if/elif/else

---

## 📖 今日讲解

条件判断是程序的 **决策中枢**，让程序根据不同情况执行不同逻辑。

基本语法：

```python
if 条件1:
    执行逻辑1
elif 条件2:
    执行逻辑2
else:
    执行逻辑3
```

条件判断的核心是 **比较运算符** 和 **逻辑运算符**：

**比较运算符**：`==`、`!=`、`>`、`<`、`>=`、`<=`

**逻辑运算符**：
- `and` —— 两个条件都满足
- `or` —— 满足一个即可
- `not` —— 取反

```python
if age >= 18 and age <= 60:
    print("适龄劳动力")
```

Python 还有个特殊写法——**条件表达式**（三元运算符）：

```python
result = "及格" if score >= 60 else "不及格"
```

---

## 💻 可运行代码示例


````python
# === 基本条件判断 ===

score = 75

if score >= 90:
    grade = "优秀"
elif score >= 80:
    grade = "良好"
elif score >= 60:
    grade = "及格"
else:
    grade = "不及格"

print(f"分数:{score}, 等级:{grade}")
````


逻辑运算符组合条件：


````python
# === 逻辑运算符 ===

age = 25
has_license = True

# and：两个条件都满足
if age >= 18 and has_license:
    print("可以开车")
else:
    print("不能开车")

# or：满足一个即可
day = "周六"
if day == "周六" or day == "周日":
    print("周末休息")

# not：取反
is_admin = False
if not is_admin:
    print("普通用户")

# 条件表达式（三元运算符）
score = 55
result = "及格" if score >= 60 else "不及格"
print(f"结果: {result}")
````


条件嵌套和多重判断：


````python
# === 条件嵌套 ===

age = 20
is_student = True

if age >= 18:
    if is_student:
        print("成年学生，半价票")
    else:
        print("成年非学生，全价票")
else:
    print("未成年，免费")

# 多条件组合
hour = 14
if hour < 12:
    print("上午")
elif hour < 18:
    print("下午")
elif hour < 22:
    print("晚上")
else:
    print("深夜")
````

# Day 7: 循环 for/while + range

---

## 📖 今日讲解

循环让程序 **重复执行** 一段代码，是编程中最核心的结构之一。

Python 有两种循环：

### for 循环（最常用）

遍历一个序列（列表、字符串、字典等）：

```python
for item in [1, 2, 3]:
    print(item)
```

配合 `range()` 函数，可以控制循环次数：

```python
for i in range(5):      # 0,1,2,3,4
    print(i)
```

### while 循环

当条件满足时一直循环：

```python
while condition:
    执行逻辑
```

循环控制语句：
- `break` —— 跳出整个循环
- `continue` —— 跳过本次，进入下一次
- `else` —— 循环正常结束（没被 break）时执行

---

## 💻 可运行代码示例


````python
# === for 循环 ===

# 遍历列表
fruits = ["苹果", "香蕉", "橘子"]
for fruit in fruits:
    print(f"我喜欢吃{fruit}")

# range() 控制循环次数
print("=== range(5) ===")
for i in range(5):
    print(i)

# range(起, 止)
print("=== range(1, 4) ===")
for i in range(1, 4):
    print(i)

# range(起, 止, 步长)
print("=== range(0, 10, 2) ===")
for i in range(0, 10, 2):
    print(i)
````


while 循环和循环控制：


````python
# === while 循环 ===

# 基本while
count = 0
while count < 3:
    print(f"第{count+1}次")
    count += 1

# break：跳出循环
print("=== break ===")
for i in range(10):
    if i == 5:
        break
    print(i)

# continue：跳过本次
print("=== continue ===")
for i in range(5):
    if i == 2:
        continue
    print(i)
````


`enumerate()` 同时获取索引和值——遍历时非常实用：


````python
# === enumerate 同时取索引和值 ===

fruits = ["苹果", "香蕉", "橘子"]

# 不用enumerate（繁琐）
i = 0
for fruit in fruits:
    print(f"{i}: {fruit}")
    i += 1

# 用enumerate（简洁，推荐）
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
````


字典遍历和嵌套循环：


````python
# === 字典遍历 ===

scores = {"张三": 90, "李四": 85, "王五": 92}
for name, score in scores.items():
    print(f"{name}: {score}分")

# 嵌套循环（九九乘法表）
print("=== 九九乘法表 ===")
for i in range(1, 4):
    for j in range(1, i + 1):
        print(f"{j}x{i}={i*j}", end=" ")
    print()  # 换行
````

# Day 8: 函数定义与参数

---

## 📖 今日讲解

函数是 **代码复用** 的核心机制。把一段常用的代码封装成函数，下次直接调用，避免重复写。

定义函数用 `def` 关键字：

```python
def 函数名(参数):
    函数体
    return 返回值
```

函数参数有四种形式（按常用程度排序）：

| 参数类型 | 示例 | 说明 |
|----------|------|------|
| 位置参数 | `def f(a, b)` | 按顺序传参 |
| 默认参数 | `def f(a, b=10)` | 有默认值，可不传 |
| 可变参数 | `def f(*args)` | 收集多余位置参数为元组 |
| 关键字参数 | `def f(**kwargs)` | 收集多余关键字参数为字典 |

> 💡 **原则**：参数顺序必须是 **位置参数 → 默认参数 → *args → **kwargs**

---

## 💻 可运行代码示例


````python
# === 基本函数定义 ===

def greet(name):
    # 打招呼函数
    return f"你好，{name}！"

# 调用函数
msg = greet("张三")
print(msg)

# 带返回值的函数
def add(a, b):
    return a + b

result = add(3, 5)
print(f"3 + 5 = {result}")
````


默认参数——让调用更灵活：


````python
# === 默认参数 ===

def create_user(name, age, city="未知"):
    return f"姓名:{name}, 年龄:{age}, 城市:{city}"

# 不传city，用默认值
print(create_user("张三", 25))

# 传city，覆盖默认值
print(create_user("李四", 30, "北京"))
````


可变参数 *args 和关键字参数 **kwargs：


````python
# === *args 可变参数 ===

def calculate_sum(*args):
    # 接收任意个数字，求和
    total = 0
    for num in args:
        total += num
    return total

# 可以传任意个参数
print(calculate_sum(1, 2))
print(calculate_sum(1, 2, 3, 4, 5))

# === **kwargs 关键字参数 ===

def create_profile(**kwargs):
    # 接收任意个关键字参数
    for key, value in kwargs.items():
        print(f"{key}: {value}")

create_profile(name="张三", age=25, city="北京", job="工程师")
````

# Day 9: 返回值与作用域

---

## 📖 今日讲解

### 返回值

函数用 `return` 返回结果。Python 函数可以 **返回多个值**（实际是返回元组）：

```python
def stats(a, b):
    return a+b, a-b, a*b

s, d, p = stats(5, 3)  # 解包
```

如果没有 `return`，函数默认返回 `None`。

### 作用域

变量的 **可见范围** 叫作用域：

- **局部变量**：函数内部定义，只在函数内有效
- **全局变量**：函数外部定义，整个文件都能访问

```python
x = 10  # 全局变量

def f():
    x = 20  # 局部变量，不影响全局
    print(x)  # 20

f()
print(x)  # 10（全局不变）
```

如果函数内要修改全局变量，用 `global` 关键字：

```python
count = 0
def increment():
    global count
    count += 1
```

> ⚠️ **建议**：尽量少用 `global`，容易引发 bug。优先通过参数传值和返回值。

---

## 💻 可运行代码示例


````python
# === 返回多个值 ===

def calculate(a, b):
    return a + b, a - b, a * b

# 元组解包接收
add, sub, mul = calculate(10, 3)
print(f"和:{add}, 差:{sub}, 积:{mul}")

# 也可以用元组接收
result = calculate(10, 3)
print(f"结果元组: {result}")

# 函数默认返回None
def say_hello():
    print("Hello")

r = say_hello()
print(f"返回值: {r}")  # None
````


作用域演示：


````python
# === 作用域 ===

# 全局变量
message = "我是全局变量"

def test_scope():
    # 局部变量
    message = "我是局部变量"
    print("函数内:", message)

test_scope()
print("函数外:", message)  # 全局变量不受影响

# global 关键字
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
increment()
print(f"counter: {counter}")  # 3
````

# Day 10: Lambda 匿名函数

---

## 📖 今日讲解

Lambda 是 **一句话的匿名函数**，适合写简单的、临时用一次的函数。

语法：

```python
lambda 参数: 表达式
```

等价于：

```python
def 函数名(参数):
    return 表达式
```

Lambda 的最大价值在于配合 `map()`、`filter()`、`sorted()` 等高阶函数使用，代码更简洁。

**三个经典用法**：

```python
# 1. sorted 排序时指定key
sorted(lst, key=lambda x: x[1])

# 2. map 对每个元素做转换
list(map(lambda x: x*2, lst))

# 3. filter 过滤
list(filter(lambda x: x > 0, lst))
```

> 💡 **建议**：复杂逻辑不要用 lambda，用普通 def 函数更清晰。Lambda 适合一行能写完的简单逻辑。

---

## 💻 可运行代码示例


````python
# === Lambda 基本用法 ===

# 普通函数
def add(a, b):
    return a + b

# 等价的lambda
add_lambda = lambda a, b: a + b

print("普通函数:", add(3, 5))
print("Lambda:", add_lambda(3, 5))

# lambda 可以直接调用
result = (lambda x: x ** 2)(5)
print("直接调用:", result)
````


Lambda 配合 sorted/map/filter——这才是 lambda 的真正用武之地：


````python
# === Lambda + sorted ===

students = [("张三", 90), ("李四", 85), ("王五", 92)]

# 按分数排序（lambda指定排序key）
sorted_by_score = sorted(students, key=lambda s: s[1])
print("按分数升序:", sorted_by_score)

# 按分数降序
sorted_desc = sorted(students, key=lambda s: s[1], reverse=True)
print("按分数降序:", sorted_desc)

# === Lambda + map ===

nums = [1, 2, 3, 4, 5]

# 每个元素乘以2
doubled = list(map(lambda x: x * 2, nums))
print("乘以2:", doubled)

# === Lambda + filter ===

# 过滤出偶数
evens = list(filter(lambda x: x % 2 == 0, nums))
print("偶数:", evens)

# 过滤出大于2的
gt2 = list(filter(lambda x: x > 2, nums))
print("大于2:", gt2)
````

# Day 12: 文件读写

## 📖 今日讲解

文件读写是 Python 最基础也是最常用的 IO 操作。

无论做爬虫（写结果到文件）、数据分析（读 CSV）、Web 开发（读配置文件），都离不开文件操作。

### 基本操作三步走

```python
# 1. 打开文件
f = open("file.txt", "w")

# 2. 读写
f.write("内容")

# 3. 关闭
f.close()
```

### 打开模式

| 模式 | 说明 |
|------|------|
| `"r"` | 读（默认） |
| `"w"` | 写（覆盖） |
| `"a"` | 追加 |
| `"r+"` | 读写 |

### 推荐用 with 语句

```python
with open("file.txt", "r") as f:
    content = f.read()
# 自动关闭，不用手动 close
```

`with` 语句会在代码块结束后 **自动关闭文件**，即使出错也不会忘记关闭。这是 Python 推荐的最佳实践。

---

## 💻 可运行代码示例


````python
# === 写文件 ===

# with语句自动管理文件关闭
with open("/tmp/test.txt", "w", encoding="utf-8") as f:
    f.write("第一行\n")
    f.write("第二行\n")
    f.write("第三行\n")

print("文件写入完成")
````


读文件的三种方式：


````python
# === 读文件 ===

# 方式1：read() 一次读完
with open("/tmp/test.txt", "r", encoding="utf-8") as f:
    content = f.read()
print("=== read() ===")
print(content)

# 方式2：readlines() 按行读取成列表
with open("/tmp/test.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
print("=== readlines() ===")
print(lines)  # 包含换行符

# 方式3：for循环逐行读（最省内存，推荐）
print("=== for循环逐行 ===")
with open("/tmp/test.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())  # strip去除换行符
````


追加内容到文件：


````python
# === 追加内容 ===

# "a" 模式追加，不会覆盖原内容
with open("/tmp/test.txt", "a", encoding="utf-8") as f:
    f.write("第四行（追加）\n")

# 验证追加结果
with open("/tmp/test.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

# Day 13: 异常处理 try/except

---

## 📖 今日讲解

程序运行时会遇到各种错误：文件不存在、网络超时、除以零、类型错误……

如果没有异常处理，程序会 **直接崩溃退出**。用 `try/except` 可以 **捕获错误并优雅处理**。

基本语法：

```python
try:
    # 可能出错的代码
    result = 10 / 0
except ZeroDivisionError:
    # 出错后的处理
    print("不能除以零")
```

### 完整结构

```python
try:
    # 可能出错的代码
except 错误类型1:
    # 处理错误1
except 错误类型2 as e:
    # 处理错误2，e 是错误信息
else:
    # 没出错时执行
finally:
    # 无论是否出错都执行（清理资源）
```

### 常见异常类型

| 异常 | 说明 |
|------|------|
| `ZeroDivisionError` | 除以零 |
| `FileNotFoundError` | 文件不存在 |
| `ValueError` | 值错误（如 int("abc")） |
| `TypeError` | 类型错误 |
| `IndexError` | 索引越界 |
| `KeyError` | 字典键不存在 |
| `AttributeError` | 属性不存在 |

---

## 💻 可运行代码示例


````python
# === 基本异常处理 ===

try:
    result = 10 / 0
except ZeroDivisionError:
    print("错误：不能除以零")

# 捕获多种异常
try:
    num = int("abc")
except ValueError:
    print("错误：无法转换为数字")
except TypeError:
    print("错误：类型不匹配")

# 获取错误信息
try:
    lst = [1, 2, 3]
    print(lst[10])
except IndexError as e:
    print(f"错误：{e}")
````


完整的 try/else/finally 结构：


````python
# === 完整结构 ===

try:
    f = open("/tmp/test.txt", "r")
    content = f.read()
except FileNotFoundError:
    print("错误：文件不存在")
else:
    print("文件读取成功")
    print(f"内容: {content[:20]}...")
finally:
    # 无论是否出错，都关闭文件
    print("finally: 清理资源")
    # f.close()  # 如果打开成功才需要关闭
````


主动抛出异常 `raise`：


````python
# === raise 主动抛出异常 ===

def check_age(age):
    if age < 0:
        raise ValueError("年龄不能为负数")
    if age > 150:
        raise ValueError("年龄不合理")
    return age

try:
    check_age(-5)
except ValueError as e:
    print(f"校验失败: {e}")

# 实际应用：参数校验
def divide(a, b):
    if b == 0:
        raise ValueError("除数不能为零")
    return a / b

try:
    divide(10, 0)
except ValueError as e:
    print(f"计算失败: {e}")
````

# Day 14: JSON 与 CSV 处理

---

## 📖 今日讲解

JSON 和 CSV 是 **数据交换最常用的两种格式**：

- **JSON**：API 接口的标准数据格式，配置文件常用
- **CSV**：表格数据的标准格式，Excel 可打开

### JSON

Python 内置 `json` 模块，四个核心函数：

| 函数 | 用途 |
|------|------|
| `json.dumps(obj)` | Python 对象 → JSON 字符串 |
| `json.loads(str)` | JSON 字符串 → Python 对象 |
| `json.dump(obj, f)` | 写入文件 |
| `json.load(f)` | 从文件读取 |

### CSV

Python 内置 `csv` 模块：

```python
import csv

# 读
with open("data.csv") as f:
    reader = csv.reader(f)

# 写
with open("out.csv", "w") as f:
    writer = csv.writer(f)
```

---

## 💻 可运行代码示例


````python
# === JSON 处理 ===

import json

# Python 字典 → JSON 字符串
data = {
    "name": "张三",
    "age": 25,
    "scores": [90, 85, 92],
    "is_student": True
}

json_str = json.dumps(data, ensure_ascii=False, indent=2)
print("=== Python → JSON ===")
print(json_str)

# JSON 字符串 → Python 字典
parsed = json.loads(json_str)
print("\n=== JSON → Python ===")
print(f"姓名: {parsed['name']}")
print(f"类型: {type(parsed)}")
````


JSON 文件读写：


````python
# === JSON 文件读写 ===

import json

# 写入JSON文件
data = {"users": [
    {"name": "张三", "age": 25},
    {"name": "李四", "age": 30}
]}

with open("/tmp/data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print("JSON文件写入完成")

# 读取JSON文件
with open("/tmp/data.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)

for user in loaded["users"]:
    print(f"{user['name']}: {user['age']}岁")
````


CSV 文件读写：


````python
# === CSV 处理 ===

import csv

# 写入CSV
rows = [
    ["姓名", "年龄", "城市"],
    ["张三", 25, "北京"],
    ["李四", 30, "上海"],
    ["王五", 28, "广州"]
]

with open("/tmp/data.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

print("CSV文件写入完成")

# 读取CSV
print("\n=== 读取CSV ===")
with open("/tmp/data.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)

# 用DictReader读取（更实用，每行是字典）
print("\n=== DictReader ===")
with open("/tmp/data.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['姓名']}, {row['年龄']}岁, {row['城市']}")
````


更完整的类——带属性修改：


````python
# === 更完整的类 ===

class BankAccount:
    # 银行账户类

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        # 存款
        self.balance += amount
        print(f"{self.owner}存入{amount}，余额: {self.balance}")

    def withdraw(self, amount):
        # 取款
        if amount > self.balance:
            print("余额不足！")
            return False
        self.balance -= amount
        print(f"{self.owner}取出{amount}，余额: {self.balance}")
        return True

    def check_balance(self):
        # 查余额
        print(f"{self.owner}当前余额: {self.balance}")

# 使用
account = BankAccount("张三", 1000)
account.deposit(500)
account.withdraw(200)
account.withdraw(5000)  # 余额不足
account.check_balance()
````

## 📖 今日讲解

```python
class Animal:          # 父类
    def eat(self):
        print("吃东西")

class Dog(Animal):     # 子类继承Animal
    def bark(self):
        print("汪汪")
```

继承的好处：**代码复用**。公共逻辑放父类，特有逻辑放子类。

### 方法重写

子类可以 **重写** 父类的方法，覆盖父类行为：

```python
class Animal:
    def sound(self):
        print("动物叫声")

class Dog(Animal):
    def sound(self):      # 重写父类方法
        print("汪汪汪")
```

### super()

用 `super()` 调用父类的方法：

```python
class Dog(Animal):
    def __init__(self, name):
        super().__init__()  # 调用父类构造方法
        self.name = name
```

---

## 💻 可运行代码示例


````python
# === 继承基础 ===

class Animal:
    # 动物基类

    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name}在吃东西")

    def sleep(self):
        print(f"{self.name}在睡觉")

class Dog(Animal):
    # 狗继承动物

    def bark(self):
        print(f"{self.name}: 汪汪汪！")

class Cat(Animal):
    # 猫继承动物

    def meow(self):
        print(f"{self.name}: 喵喵喵~")

# Dog 和 Cat 都继承了 Animal 的 eat/sleep 方法
dog = Dog("旺财")
dog.eat()      # 继承自Animal
dog.sleep()    # 继承自Animal
dog.bark()     # 自己的方法

cat = Cat("咪咪")
cat.eat()      # 继承自Animal
cat.meow()     # 自己的方法
````


方法重写和 super()：


````python
# === 方法重写 ===

class Animal:
    def __init__(self, name):
        self.name = name

    def sound(self):
        print(f"{self.name}: ...")

    def info(self):
        print(f"我是动物，叫{self.name}")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # 调用父类构造方法
        self.breed = breed      # 子类新增属性

    def sound(self):            # 重写父类方法
        print(f"{self.name}: 汪汪汪！")

    def info(self):             # 重写并扩展
        super().info()          # 先调用父类的info
        print(f"品种是{self.breed}")

dog = Dog("旺财", "金毛")
dog.sound()
dog.info()
````

# Day 17: 类的常用魔术方法

---

## 📖 今日讲解

**魔术方法** 是 Python 类中以双下划线开头和结尾的方法（也叫 dunder method）。

它们让自定义类能像内置类型一样工作。

最常用的魔术方法：

| 方法 | 触发时机 | 用途 |
|------|----------|------|
| `__init__` | 创建对象时 | 初始化属性 |
| `__str__` | `print(obj)` / `str(obj)` | 控制打印输出 |
| `__repr__` | 调试时显示 | 开发者友好的表示 |
| `__len__` | `len(obj)` | 支持len() |
| `__eq__` | `obj1 == obj2` | 支持==比较 |
| `__lt__` | `obj1 < obj2` | 支持<比较 |

其中 `__str__` 最常用——让 print 输出自定义对象时更有意义，而不是默认的 `<__main__.XXX object at 0x...>`。

---

## 💻 可运行代码示例


````python
# === __str__ 和 __repr__ ===

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        # print时显示，面向用户
        return f"商品:{self.name}, 价格:{self.price}元"

    def __repr__(self):
        # 调试时显示，面向开发者
        return f"Product(name='{self.name}', price={self.price})"

p = Product("iPhone", 5999)

# print 触发 __str__
print(p)

# 直接在交互式环境触发 __repr__
print(repr(p))
````


`__len__`、`__eq__` 等比较方法：


````python
# === __len__ 和比较方法 ===

class Team:
    def __init__(self, name, members):
        self.name = name
        self.members = members

    def __len__(self):
        # 支持 len() 函数
        return len(self.members)

    def __str__(self):
        return f"团队:{self.name}({len(self.members)}人)"

    def __eq__(self, other):
        # 支持 == 比较
        return len(self.members) == len(other.members)

    def __lt__(self, other):
        # 支持 < 比较
        return len(self.members) < len(other.members)

team1 = Team("开发组", ["张三", "李四", "王五"])
team2 = Team("测试组", ["赵六", "钱七"])

print(team1)
print(f"团队人数: {len(team1)}")

print(f"两个团队人数相等: {team1 == team2}")
print(f"team1 < team2: {team1 < team2}")
````

# Day 18: 推导式进阶

---

## 📖 今日讲解

**推导式（Comprehension）** 是 Python 的特色语法，用一行代码创建列表/字典/集合，比循环更简洁高效。

四种推导式：

```python
# 列表推导式（最常用）
[x for x in range(10)]

# 带条件
[x for x in range(10) if x % 2 == 0]

# 带转换
[x**2 for x in range(5)]

# 字典推导式
{k: v for k, v in items}

# 集合推导式
{x for x in nums}
```

基本语法模板：

```python
[表达式 for 变量 in 可迭代对象 if 条件]
```

执行逻辑：遍历可迭代对象 → 过滤条件 → 对每个元素应用表达式 → 收集到列表。

> 💡 **性能**：推导式比 for 循环 + append 快约 30%，Python 内部做了优化。

---

## 💻 可运行代码示例


````python
# === 列表推导式 ===

# 基本用法：生成平方数列表
squares = [x**2 for x in range(1, 6)]
print("平方数:", squares)

# 带条件过滤
evens = [x for x in range(1, 11) if x % 2 == 0]
print("偶数:", evens)

# 带转换
names = ["alice", "bob", "charlie"]
upper_names = [name.capitalize() for name in names]
print("首字母大写:", upper_names)

# 嵌套循环
pairs = [(x, y) for x in range(2) for y in range(2)]
print("组合:", pairs)
````


字典推导式和集合推导式：


````python
# === 字典推导式 ===

# 列表转字典（带转换）
fruits = ["苹果", "香蕉", "橘子"]
price_dict = {fruit: len(fruit) for fruit in fruits}
print("水果长度:", price_dict)

# 反转字典
original = {"a": 1, "b": 2, "c": 3}
reversed_dict = {v: k for k, v in original.items()}
print("反转:", reversed_dict)

# 带条件过滤
scores = {"张三": 85, "李四": 55, "王五": 92}
passed = {name: score for name, score in scores.items() if score >= 60}
print("及格:", passed)

# === 集合推导式 ===
nums = [1, 2, 2, 3, 3, 3, 4]
unique_squares = {x**2 for x in nums}
print("去重平方:", unique_squares)
````

# Day 19: 迭代器与生成器

---

## 📖 今日讲解

**生成器（Generator）** 是一种特殊的函数，用 `yield` 代替 `return` 返回数据。

普通函数一次返回所有结果，**生成器按需一次返回一个**，节省内存。

```python
def counter():
    yield 1
    yield 2
    yield 3
```

### 为什么用生成器？

处理大数据时，普通方式会一次性把所有数据加载到内存：

```python
# 普通方式：100万个数字全部加载到内存
nums = [x for x in range(1000000)]

# 生成器：一次只生成一个，几乎不占内存
nums = (x for x in range(1000000))
```

### 两种创建方式

```python
# 方式1：生成器表达式（类似列表推导式，用圆括号）
gen = (x**2 for x in range(5))

# 方式2：生成器函数（用yield）
def gen_func():
    yield 1
    yield 2
```

---

## 💻 可运行代码示例


````python
# === 生成器函数 ===

def counter():
    # 生成器函数
    yield "第一"
    yield "第二"
    yield "第三"

# 使用生成器
gen = counter()
print(next(gen))  # 第一
print(next(gen))  # 第二
print(next(gen))  # 第三

# 用for遍历（推荐，不会StopIteration）
for item in counter():
    print(item)
````


生成器表达式——处理大数据利器：

# Day 20: 装饰器入门

---

## 📖 今日讲解

**装饰器（Decorator）** 是 Python 的高级特性，用来 **在不修改原函数的前提下，给函数增加功能**。

语法用 `@` 符号：

```python
@my_decorator
def hello():
    print("hello")
```

等价于：

```python
def hello():
    print("hello")
hello = my_decorator(hello)
```

### 装饰器的本质

装饰器是一个 **接收函数、返回函数** 的函数：

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        # 执行前做的事
        result = func(*args, **kwargs)
        # 执行后做的事
        return result
    return wrapper
```

### 最常见用途

- **计时**：测量函数执行时间
- **日志**：记录函数调用
- **权限校验**：检查用户是否登录
- **缓存**：缓存函数结果

---

## 💻 可运行代码示例


````python
# === 基本装饰器 ===

def log_decorator(func):
    # 日志装饰器：在函数调用前后打印日志
    def wrapper(*args, **kwargs):
        print(f"=== 调用 {func.__name__} ===")
        result = func(*args, **kwargs)
        print(f"=== {func.__name__} 执行完毕 ===")
        return result
    return wrapper

@log_decorator
def greet(name):
    print(f"你好，{name}！")

# 调用被装饰的函数
greet("张三")
````


实用的计时装饰器：


````python
# === 计时装饰器 ===

import time

def timer(func):
    # 测量函数执行时间
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} 耗时: {end - start:.4f}秒")
        return result
    return wrapper

@timer
def slow_function():
    # 模拟耗时操作
    total = 0
    for i in range(1000000):
        total += i
    return total

result = slow_function()
print(f"结果: {result}")
````


带参数的装饰器——更灵活：


````python
# === 带参数的装饰器 ===

def repeat(n):
    # 让函数重复执行n次
    def decorator(func):
        def wrapper(*args, **kwargs):
            for i in range(n):
                print(f"--- 第{i+1}次执行 ---")
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(3)  # 重复执行3次
def say_hi():
    print("Hi!")

say_hi()
````

## 💻 可运行代码示例


````python
# === 基本日期时间 ===

from datetime import datetime, timedelta

# 当前时间
now = datetime.now()
print("当前时间:", now)

# 格式化输出
formatted = now.strftime("%Y年%m月%d日 %H:%M:%S")
print("格式化:", formatted)

# 字符串转日期
date_str = "2024-01-15 10:30:00"
parsed = datetime.strptime(date_str, "%Y-%m-%d %H:%M:%S")
print("字符串转日期:", parsed)
print("类型:", type(parsed))
````


时间加减和差值：


````python
# === 时间加减 ===

from datetime import datetime, timedelta

now = datetime.now()

# 未来时间
future_7days = now + timedelta(days=7)
print(f"7天后: {future_7days.strftime('%Y-%m-%d')}")

future_1hour = now + timedelta(hours=1)
print(f"1小时后: {future_1hour.strftime('%H:%M')}")

# 过去时间
past_30days = now - timedelta(days=30)
print(f"30天前: {past_30days.strftime('%Y-%m-%d')}")

# 时间差
date1 = datetime(2024, 1, 1)
date2 = datetime(2024, 12, 31)
diff = date2 - date1
print(f"相差: {diff.days}天")
````


日期比较和实用操作：


````python
# === 日期比较 ===

from datetime import datetime

date1 = datetime(2024, 1, 15)
date2 = datetime(2024, 6, 1)

# 比较
print(f"date1 < date2: {date1 < date2}")
print(f"date1 == date2: {date1 == date2}")

# 判断是否周末
def is_weekend(dt):
    return dt.weekday() >= 5  # 周六=5, 周日=6

today = datetime.now()
print(f"今天是否周末: {is_weekend(today)}")

# 计算年龄
def calculate_age(birth_date):
    today = datetime.now()
    age = today.year - birth_date.year
    # 如果今年生日还没到，减1
    if (today.month, today.day) < (birth_date.month, birth_date.day):
        age -= 1
    return age

birth = datetime(1999, 5, 20)
print(f"年龄: {calculate_age(birth)}岁")
````

# Day 22: 正则表达式 re

---

## 📖 今日讲解

**正则表达式** 是文本处理的利器，用于 **匹配、查找、替换** 字符串中的模式。

Python 内置 `re` 模块提供正则支持。

核心函数：

| 函数 | 用途 |
|------|------|
| `re.match()` | 从开头匹配 |
| `re.search()` | 搜索任意位置（返回第一个） |
| `re.findall()` | 找出所有匹配 |
| `re.sub()` | 替换匹配内容 |
| `re.split()` | 按模式分割 |

常用正则符号：

| 符号 | 含义 | 示例 |
|------|------|------|
| `\d` | 数字 | `\d+` 匹配 123 |
| `\w` | 字母数字下划线 | `\w+` |
| `\s` | 空白字符 | |
| `.` | 任意字符 | |
| `*` | 0次或多次 | `a*` |
| `+` | 1次或多次 | `a+` |
| `?` | 0次或1次 | `a?` |
| `{n,m}` | n到m次 | `\d{2,4}` |
| `^` / `$` | 开头 / 结尾 | |
| `[...]` | 字符集 | `[a-z]` |

---

## 💻 可运行代码示例


````python
# === 基本匹配 ===

import re

# re.search 搜索任意位置
text = "我的手机号是13812345678，邮箱是test@example.com"

# 找手机号（1开头的11位数字）
phone = re.search(r"1\d{10}", text)
if phone:
    print("手机号:", phone.group())

# re.findall 找出所有匹配
numbers = re.findall(r"\d+", text)
print("所有数字:", numbers)

# re.match 从开头匹配
result = re.match(r"hello", "hello world")
if result:
    print("match成功:", result.group())
````


分组提取和替换：


````python
# === 分组和替换 ===

import re

# 分组提取
text = "2024-01-15 发布"
pattern = r"(\d{4})-(\d{2})-(\d{2})"
match = re.search(pattern, text)
if match:
    print(f"年: {match.group(1)}")
    print(f"月: {match.group(2)}")
    print(f"日: {match.group(3)}")
    print(f"完整: {match.group(0)}")

# re.sub 替换
# 隐藏手机号中间4位
phone_text = "联系我: 13812345678"
hidden = re.sub(r"(1\d{2})\d{4}(\d{4})", r"\1****\2", phone_text)
print("隐藏后:", hidden)

# re.split 分割
parts = re.split(r"[,;|]", "a,b;c|d")
print("分割结果:", parts)
````


常见正则验证：


````python
# === 常见验证 ===

import re

# 验证邮箱
def is_valid_email(email):
    pattern = r"^[\w.-]+@[\w.-]+\.\w+$"
    return bool(re.match(pattern, email))

print(is_valid_email("test@example.com"))   # True
print(is_valid_email("invalid-email"))      # False

# 验证手机号
def is_valid_phone(phone):
    pattern = r"^1[3-9]\d{9}$"
    return bool(re.match(pattern, phone))

print(is_valid_phone("13812345678"))  # True
print(is_valid_phone("123456"))       # False

# 提取URL
text = "访问 https://www.python.org 和 http://github.com"
urls = re.findall(r"https?://[\w./]+", text)
print("URL:", urls)
````

## 📖 今日讲解

**虚拟环境** 是 Python 项目的隔离容器，每个项目有自己独立的依赖包版本，互不干扰。

### 为什么需要虚拟环境？

假设项目 A 需要 Django 3.x，项目 B 需要 Django 4.x。如果都装在全局环境，版本会冲突。虚拟环境解决的就是这个问题。

### 创建虚拟环境

```bash
# 创建虚拟环境（名为 venv）
python -m venv venv

# 激活（Linux/Mac）
source venv/bin/activate

# 激活（Windows）
venv\Scripts\activate

# 退出虚拟环境
deactivate
```

### 依赖管理

```bash
# 安装包
pip install requests

# 导出依赖列表
pip freeze > requirements.txt

# 根据依赖列表安装
pip install -r requirements.txt
```

`requirements.txt` 是项目的依赖清单，别人拿到你的代码后，一条命令就能安装所有依赖。

---

## 💻 可运行代码示例

> ⚠️ 虚拟环境是命令行操作，这里演示依赖管理的代码部分：


````python
# === 依赖管理演示 ===

# 模拟生成 requirements.txt
dependencies = [
    "requests==2.31.0",
    "flask==3.0.0",
    "pandas==2.1.0",
    "numpy==1.24.0"
]

# 写入 requirements.txt
with open("/tmp/requirements.txt", "w") as f:
    for dep in dependencies:
        f.write(dep + "\n")

print("requirements.txt 已生成")

# 读取并解析
print("\n=== 依赖列表 ===")
with open("/tmp/requirements.txt", "r") as f:
    for line in f:
        line = line.strip()
        if line:
            name, version = line.split("==")
            print(f"包名: {name}, 版本: {version}")
````


`__name__ == "__main__"` 的用法——Python 项目的入口惯例：


````python
# === __name__ == "__main__" ===

# 这个知识点在项目结构中非常重要

def main():
    # 主函数
    print("程序启动")
    print("执行主要逻辑...")

def helper():
    # 辅助函数
    print("辅助功能")

# 当直接运行此文件时执行main
# 被导入时不执行
if __name__ == "__main__":
    main()

# 解释：
# 直接运行本文件时，__name__ 的值是 "__main__"
# 被别的文件 import 时，__name__ 是模块名（如 "my_module"）
# 这样设计可以避免导入时自动执行不该执行的代码
````

# Day 25: requests 网络请求

## 📖 今日讲解

| 方法 | 用途 |

```

### Response 对象

```

## 💻 可运行代码示例

> 注：以下代码需要联网，如果网络不通可以看注释理解


````python
# === GET 请求 ===

import requests

# 访问公开API（获取GitHub用户信息）
try:
    resp = requests.get(
        "https://api.github.com/users/python",
        timeout=10
    )

    print(f"状态码: {resp.status_code}")

    if resp.status_code == 200:
        data = resp.json()
        print(f"用户名: {data['login']}")
        print(f"粉丝数: {data['followers']}")
        print(f"简介: {data.get('bio', '无')}")
except Exception as e:
    print(f"请求失败: {e}")
````


带参数和请求头的请求：


````python
# === 带参数和请求头 ===

import requests

# 带查询参数
try:
    resp = requests.get(
        "https://api.github.com/search/repositories",
        params={
            "q": "python",
            "sort": "stars",
            "per_page": 3
        },
        headers={
            "Accept": "application/vnd.github.v3+json"
        },
        timeout=10
    )

    if resp.status_code == 200:
        data = resp.json()
        print(f"总数: {data['total_count']}")
        for repo in data.get("items", []):
            print(f"  {repo['full_name']}: {repo['stargazers_count']}星")
except Exception as e:
    print(f"请求失败: {e}")
````


POST 请求和异常处理：


````python
# === 安全的请求函数 ===

import requests

def safe_request(url, method="GET", **kwargs):
    # 带异常处理的安全请求
    try:
        resp = requests.request(method, url, timeout=10, **kwargs)
        resp.raise_for_status()  # 状态码非200抛异常
        return resp
    except requests.Timeout:
        print("请求超时")
    except requests.ConnectionError:
        print("连接失败")
    except requests.HTTPError as e:
        print(f"HTTP错误: {e}")
    except Exception as e:
        print(f"其他错误: {e}")
    return None

# 使用
resp = safe_request("https://httpbin.org/get")
if resp:
    print("请求成功!")
    print(resp.json())
````

## 📖 今日讲解

`print()` 虽然简单，但在正式项目中 **不够用**：

### 日志级别（从低到高）

| 级别 | 数字 | 用途 |
|------|------|------|
| DEBUG | 10 | 调试信息 |
| INFO | 20 | 正常运行信息 |
| WARNING | 30 | 警告 |
| ERROR | 40 | 错误 |
| CRITICAL | 50 | 严重错误 |

### 基本用法

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filename="app.log"
)

logging.info("程序启动")
logging.error("发生错误")
```

---

## 💻 可运行代码示例


````python
# === 基本日志 ===

import logging

# 配置日志
logging.basicConfig(
    level=logging.DEBUG,  # 设置最低显示级别
    format="%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    filename="/tmp/app.log",
    filemode="w"  # w覆盖, a追加
)

# 不同级别日志
logging.debug("这是调试信息")
logging.info("程序正常启动")
logging.warning("磁盘空间不足")
logging.error("数据库连接失败")
logging.critical("系统崩溃！")

print("日志已写入 /tmp/app.log")

# 查看日志内容
with open("/tmp/app.log", "r") as f:
    print("\n=== 日志内容 ===")
    print(f.read())
````

````python
# === 同时输出到文件和控制台 ===

import logging

# 创建logger
logger = logging.getLogger("my_app")
logger.setLevel(logging.DEBUG)

# 文件handler（记录所有级别）
file_handler = logging.FileHandler("/tmp/app2.log")
file_handler.setLevel(logging.DEBUG)

# 控制台handler（只显示WARNING及以上）
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.WARNING)

# 格式
fmt = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")
file_handler.setFormatter(fmt)
console_handler.setFormatter(fmt)

# 添加handler
logger.addHandler(file_handler)
logger.addHandler(console_handler)

# 使用
logger.debug("调试信息（只写文件）")
logger.info("普通信息（只写文件）")
logger.warning("警告（文件+控制台）")
logger.error("错误（文件+控制台）")
````

````python
# === 单元测试 ===

import unittest

# 把上面的函数定义复制到这里（Jupyter中实际使用时）
def add(a, b):
    return a + b

def is_even(n):
    return n % 2 == 0

def get_grade(score):
    if score >= 90: return "A"
    elif score >= 80: return "B"
    elif score >= 60: return "C"
    else: return "D"

def divide(a, b):
    if b == 0:
        raise ValueError("除数不能为零")
    return a / b


class TestMathFunctions(unittest.TestCase):

    def setUp(self):
        # 每个测试前执行
        print("准备测试...")

    def tearDown(self):
        # 每个测试后执行
        print("测试完成\n")

    def test_add(self):
        # 测试加法
        self.assertEqual(add(1, 2), 3)
        self.assertEqual(add(-1, 1), 0)
        self.assertEqual(add(0, 0), 0)

    def test_is_even(self):
        # 测试偶数判断
        self.assertTrue(is_even(4))
        self.assertFalse(is_even(5))
        self.assertTrue(is_even(0))

    def test_get_grade(self):
        # 测试等级转换
        self.assertEqual(get_grade(95), "A")
        self.assertEqual(get_grade(85), "B")
        self.assertEqual(get_grade(65), "C")
        self.assertEqual(get_grade(55), "D")

    def test_divide(self):
        # 测试除法
        self.assertEqual(divide(10, 2), 5)
        # 测试是否抛出异常
        with self.assertRaises(ValueError):
            divide(10, 0)


# 运行测试
print("=" * 40)
print("运行单元测试")
print("=" * 40)
unittest.main(argv=[""], exit=False, verbosity=2)
````

## 📖 今日讲解

### 项目：个人记账本系统

功能：

2. 查看所有记录
3. 统计收支汇总
4. 数据保存到 JSON 文件
5. 数据可视化

### 用到的知识点

| 知识点 | 用在哪 |
|--------|--------|
| 类与对象 | 记账本类 |
| 字典/列表 | 存储记录 |
| 文件读写 | 保存数据 |

---

## 💻 可运行代码示例


````python
# === 个人记账本系统 ===

import json
import logging
from datetime import datetime
from pathlib import Path

# 配置日志
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filename="/tmp/account.log"
)

class AccountBook:
    # 记账本类

    def __init__(self, data_file="/tmp/account_data.json"):
        self.data_file = Path(data_file)
        self.records = self._load_data()
        logging.info("记账本初始化完成")

    def _load_data(self):
        # 从文件加载数据
        try:
            if self.data_file.exists():
                return json.loads(self.data_file.read_text(encoding="utf-8"))
        except Exception as e:
            logging.error(f"加载数据失败: {e}")
        return []

    def _save_data(self):
        # 保存数据到文件
        try:
            self.data_file.write_text(
                json.dumps(self.records, ensure_ascii=False, indent=2),
                encoding="utf-8"
            )
        except Exception as e:
            logging.error(f"保存数据失败: {e}")

    def add_record(self, record_type, amount, category, note=""):
        # 添加一条记录
        try:
            amount = float(amount)
            if amount <= 0:
                raise ValueError("金额必须大于0")

            record = {
                "date": datetime.now().strftime("%Y-%m-%d %H:%M"),
                "type": record_type,    # income 或 expense
                "amount": amount,
                "category": category,
                "note": note
            }
            self.records.append(record)
            self._save_data()
            logging.info(f"添加记录: {record}")
            print(f"✅ 记录已添加: {record_type} {amount}元 - {category}")
        except ValueError as e:
            print(f"❌ 输入错误: {e}")
            logging.error(f"添加记录失败: {e}")

    def list_records(self):
        # 查看所有记录
        if not self.records:
            print("暂无记录")
            return

        print("\n" + "=" * 60)
        print(f"{'日期':<20} {'类型':<8} {'金额':<10} {'分类':<10} {'备注'}")
        print("-" * 60)
        for r in self.records:
            type_label = "收入" if r["type"] == "income" else "支出"
            print(f"{r['date']:<20} {type_label:<8} {r['amount']:<10} {r['category']:<10} {r['note']}")
        print("=" * 60)

    def get_summary(self):
        # 统计汇总
        if not self.records:
            print("暂无记录")
            return

        # 用推导式分别计算收入和支出
        incomes = [r["amount"] for r in self.records if r["type"] == "income"]
        expenses = [r["amount"] for r in self.records if r["type"] == "expense"]

        total_income = sum(incomes)
        total_expense = sum(expenses)
        balance = total_income - total_expense

        print("\n📊 收支汇总")
        print(f"总收入: {total_income:.2f}元")
        print(f"总支出: {total_expense:.2f}元")
        print(f"结余: {balance:.2f}元")

        # 按分类统计支出
        from collections import defaultdict
        category_stats = defaultdict(float)
        for r in self.records:
            if r["type"] == "expense":
                category_stats[r["category"]] += r["amount"]

        if category_stats:
            print("\n支出分类:")
            for cat, amt in sorted(category_stats.items(), key=lambda x: x[1], reverse=True):
                print(f"  {cat}: {amt:.2f}元")

        return total_income, total_expense, balance


# === 使用记账本 ===
book = AccountBook()

# 添加记录
book.add_record("income", 8000, "工资", "1月工资")
book.add_record("expense", 1500, "餐饮", "1月伙食")
book.add_record("expense", 800, "交通", "地铁公交")
book.add_record("income", 500, "兼职", "副业收入")
book.add_record("expense", 2000, "房租", "1月房租")
book.add_record("expense", 300, "娱乐", "看电影")
````


查看记录和统计汇总：


````python
# === 查看所有记录 ===
book.list_records()

# === 统计汇总 ===
book.get_summary()
````

### Day 13 对应：异常处理

**来源：Flask** (`src/flask/app.py`)

Flask 的 WSGI 请求处理，展示了 **嵌套 try/except/finally** 的经典模式：



````python
# === 来自 Flask 源码（简化） ===
# 文件: src/flask/app.py
# 功能: WSGI 请求处理主循环

error = None
response = None

try:
    try:
        # 正常处理请求
        print("处理请求中...")
        response = {"status": "ok"}
    except ValueError as e:
        # 已知异常: 值错误
        error = e
        response = {"status": "error", "msg": "参数错误"}
    except Exception as e:
        # 未知异常: 重新抛出
        error = e
        raise
    print(f"响应: {response}")
finally:
    # 无论是否出错，都要清理上下文
    print("清理请求上下文")
````


**来源：FastAPI** (`fastapi/routing.py`)

FastAPI 用 `raise ... from e` 保持 **异常链**：



````python
# === 来自 FastAPI 源码（简化） ===
# 文件: fastapi/routing.py
# 功能: 请求体解析异常处理

import json

def parse_body(body_bytes):
    try:
        return json.loads(body_bytes)
    except json.JSONDecodeError as e:
        # raise from e: 保留原始异常链
        # 这样 traceback 会同时显示 JSONDecodeError 和 ValueError
        raise ValueError("请求体不是有效的JSON") from e

try:
    parse_body(b"invalid json")
except ValueError as e:
    print(f"捕获到: {e}")
    print(f"原始异常: {e.__cause__}")  # JSONDecodeError
````


**讲解**：

1. **嵌套 try/except/finally** — 外层 finally 保证资源清理，内层 except 区分已知/未知异常
2. **`raise ... from e`** — 抛出新异常时保留原始异常链。`e.__cause__` 可以访问原始异常

这两个模式在 **Web 框架、数据库驱动、网络库** 中随处可见，是健壮代码的标志。

Flask 的核心类继承：



````python
# === 来自 Flask 源码（简化） ===
# 文件: src/flask/app.py
# 功能: Flask 类继承自 App 基类

class App:
    # 基类: 定义通用 Web 应用逻辑
    def __init__(self, import_name):
        self.import_name = import_name
        self.config = {"DEBUG": False}

    def run(self):
        print(f"启动应用: {self.import_name}")

class Flask(App):
    # 子类: 扩展 WSGI 相关功能
    def __init__(self, import_name, static_folder="static"):
        super().__init__(import_name)  # 调用父类初始化
        self.static_folder = static_folder

    def __call__(self, environ, start_response):
        # 让实例可调用（WSGI 协议要求）
        return self.wsgi_app(environ, start_response)

    def wsgi_app(self, environ, start_response):
        return [b"Hello World"]

# 使用
app = Flask("myapp")
app.run()
print(f"静态目录: {app.static_folder}")
print(f"配置: {app.config}")
````


**讲解**：

1. **NamedTuple** — 当你需要一个简单的数据容器时，比完整 class 轻量，比普通 tuple 可读
2. **类继承 + super()** — Flask 继承 App 基类，复用通用逻辑，扩展特有功能

### Day 16 对应：继承

**来源：FastAPI** (`fastapi/routing.py`)

FastAPI 的 `APIRoute` 继承自 Starlette 的 `Route`，并扩展了大量功能：



````python
# === 来自 FastAPI 源码（简化） ===
# 文件: fastapi/routing.py
# 功能: 扩展 Starlette 的路由类

# Starlette 的基础路由类
class Route:
    def __init__(self, path, endpoint):
        self.path = path
        self.endpoint = endpoint

    def matches(self, scope):
        # 基础路由匹配逻辑
        return True, scope

# FastAPI 继承并扩展
class APIRoute(Route):
    # 新增类属性
    response_model = None
    status_code = 200
    summary = None

    def __init__(self, path, endpoint, *, response_model=None, **kwargs):
        super().__init__(path, endpoint)  # 调用父类构造
        self.response_model = response_model  # 新增属性
        self.app = self.get_route_handler()

    def get_route_handler(self):
        # 重写父类行为
        def handler(request):
            return {"path": self.path}
        return handler

# 使用
route = APIRoute("/users", lambda: None, response_model=dict)
print(f"路径: {route.path}")
print(f"响应模型: {route.response_model}")
print(f"状态码: {route.status_code}")
````


**讲解**：

FastAPI 的继承模式是 **框架扩展的经典范式**：

1. **继承第三方库的类**（Starlette 的 Route）
2. **保留父类功能**（super().__init__）
3. **新增属性和方法**（response_model、get_route_handler）
4. **重写父类行为**（matches、get_route_handler）

这种模式让你在 **不修改原库代码** 的前提下扩展功能，是开闭原则的体现。

### Day 17 对应：魔术方法

**来源：pytest** (`src/_pytest/main.py`)

pytest 的 Session 类用 `__repr__` 提供调试友好的输出：



````python
# === 来自 pytest 源码（简化） ===
# 文件: src/_pytest/main.py
# 功能: Session 类的调试输出

class Session:
    def __init__(self, name="", testsfailed=0, testscollected=0):
        self.name = name
        self.testsfailed = testsfailed
        self.testscollected = testscollected
        self.exitstatus = "<UNSET>"

    def __repr__(self):
        # % 格式化 + f-string 混用
        return (
            f"<{self.__class__.__name__} {self.name} "
            f"exitstatus=%r "
            f"testsfailed={self.testsfailed} "
            f"testscollected={self.testscollected}>"
        ) % getattr(self, "exitstatus", "<UNSET>")

# 调试时直接打印对象就能看到有用信息
session = Session("test_session", testsfailed=2, testscollected=10)
print(session)  # 自动调用 __repr__
print(repr(session))
````


**来源：rich** (`rich/console.py`)

rich 的 Console 类实现了 `__repr__` 和上下文管理器 `__enter__/__exit__`：



````python
# === 来自 rich 源码（简化） ===
# 文件: rich/console.py
# 功能: Console 类的魔术方法

class Console:
    def __init__(self, width=80):
        self._width = width
        self._buffer = []

    def __repr__(self):
        # 简洁的调试输出
        return f"<console width={self._width}>"

    def __enter__(self):
        # 进入 with 块: 开始缓冲
        self._buffer = []
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        # 退出 with 块: 输出缓冲内容
        print("".join(self._buffer))
        self._buffer = []

    def print(self, text):
        self._buffer.append(text)

# 使用 with 语句
with Console(width=120) as console:
    console.print("Hello, ")
    console.print("Rich!")
# 退出 with 块时自动输出
````


**讲解**：

1. **`__repr__`** — 让 `print(obj)` 和调试时显示有意义的信息，而不是 `<__main__.XXX object at 0x...>`
2. **`__enter__` / `__exit__`** — 实现 **上下文管理器**，让对象支持 `with` 语句。`__exit__` 的三个参数接收异常信息，可以处理或抑制异常

这两个魔术方法是 **让自定义类更好用** 的关键。

### Day 19 对应：生成器

**来源：pytest** (`src/_pytest/main.py`)

pytest 的测试收集器用 **生成器** 递归收集测试：



````python
# === 来自 pytest 源码（简化） ===
# 文件: src/_pytest/main.py
# 功能: 递归收集测试节点

# 模拟测试目录结构
test_tree = {
    "name": "tests",
    "children": [
        {"name": "test_auth.py", "children": [
            {"name": "test_login"},
            {"name": "test_logout"}
        ]},
        {"name": "test_api.py", "children": [
            {"name": "test_get"},
            {"name": "test_post"}
        ]}
    ]
}

# pytest 用生成器递归收集
def collect(node):
    # 如果是测试用例，直接 yield
    if "children" not in node:
        yield node["name"]
    else:
        # 如果是目录/文件，递归收集子节点
        for child in node["children"]:
            yield from collect(child)  # yield from 委托给子生成器

# 收集所有测试
for test_name in collect(test_tree):
    print(f"收集到: {test_name}")
````


**讲解**：

pytest 用生成器收集测试的两个关键点：

1. **`yield`** — 每找到一个测试就生成一个，不需要一次性把所有结果存到内存
2. **`yield from`** — 委托给子生成器，实现 **递归生成器**。`yield from collect(child)` 等价于 `for x in collect(child): yield x`

这种模式在 **遍历树形结构**（文件系统、DOM 树、AST）时非常实用。

### Day 20 对应：装饰器

**来源：Flask** (`src/flask/app.py`)

Flask 用 `functools.update_wrapper` 保留被装饰函数的元信息：



````python
# === 来自 Flask 源码 ===
# 文件: src/flask/app.py
# 功能: 上下文处理装饰器

from functools import update_wrapper

def remove_ctx(f):
    # 装饰器: 移除第一个参数如果是 AppContext
    def wrapper(self, *args, **kwargs):
        if args and isinstance(args[0], str):  # 简化判断
            args = args[1:]  # 移除第一个参数
        return f(self, *args, **kwargs)

    # 关键: 保留原函数的元信息
    update_wrapper(wrapper, f)
    return wrapper

# 使用
@remove_ctx
def handle_request(app, path):
    """处理HTTP请求"""
    return f"处理: {path}"

print(f"函数名: {handle_request.__name__}")  # handle_request（不是wrapper）
print(f"文档: {handle_request.__doc__}")     # 处理HTTP请求
````


**来源：rich** (`rich/console.py`)

rich 用 `@wraps` 装饰器（`update_wrapper` 的语法糖）实现自定义装饰器工厂：



````python
# === 来自 rich 源码 ===
# 文件: rich/console.py
# 功能: renderable group 装饰器工厂

from functools import wraps

def group(fit=True):
    # 带参数的装饰器工厂
    def decorator(method):
        @wraps(method)  # 保留原函数元信息
        def _replace(*args, **kwargs):
            renderables = method(*args, **kwargs)
            return {"items": list(renderables), "fit": fit}
        return _replace
    return decorator

# 使用
@group(fit=False)
def my_renderables():
    """生成渲染对象"""
    yield "item1"
    yield "item2"

result = my_renderables()
print(f"结果: {result}")
print(f"函数名: {my_renderables.__name__}")  # my_renderables
print(f"文档: {my_renderables.__doc__}")     # 生成渲染对象
````


**讲解**：

1. **`update_wrapper` / `@wraps`** — 装饰器的最佳实践。没有它，被装饰函数的 `__name__`、`__doc__` 会变成 wrapper 的信息
2. **带参数的装饰器工厂** — `@group(fit=False)` 这种用法，是三层嵌套函数：工厂 → 装饰器 → 包装器

Flask 和 rich 的源码中，几乎所有装饰器都用了 `@wraps`，这是 **专业代码的标志**。

### Day 22 对应：正则表达式

**来源：FastAPI** (`fastapi/routing.py`)

FastAPI 在处理路径时需要解析 URL 参数：



````python
# === 来自 FastAPI 模式（简化） ===
# 功能: 解析 URL 路径中的参数

import re

# FastAPI 路由路径: /users/{user_id}/posts/{post_id}
path = "/users/{user_id}/posts/{post_id}"

# 用正则提取路径参数
param_pattern = re.compile(r"\{(\w+)\}")
params = param_pattern.findall(path)
print(f"路径参数: {params}")  # ['user_id', 'post_id']

# 将路径转为正则模式
regex = param_pattern.sub(r"(?P<\1>[^/]+)", path)
print(f"正则模式: ^{regex}$")

# 匹配实际URL
match = re.match(f"^{regex}$", "/users/123/posts/456")
if match:
    print(f"user_id: {match.group('user_id')}")
    print(f"post_id: {match.group('post_id')}")
````


**讲解**：

FastAPI 用正则表达式实现 **路径参数提取**：

1. `re.compile` 预编译正则（性能优化，重复使用时更快）
2. `findall` 提取所有参数名
3. `sub` 把 `{param}` 替换成命名捕获组 `(?P<param>[^/]+)`
4. `match.group('name')` 按名称获取捕获结果

这个模式在 **所有 Web 框架** 的路由系统中都能看到。

### Day 29 对应：单元测试

**来源：pytest** (`src/_pytest/main.py`)

pytest 自身的测试收集逻辑就是 **最佳实践参考**：



````python
# === 来自 pytest 源码（简化） ===
# 文件: src/_pytest/main.py
# 功能: 测试收集和执行

import unittest

# pytest 的核心测试循环
class TestRunner:
    def __init__(self):
        self.testsfailed = 0
        self.testscollected = 0

    def pytest_runtestloop(self, session_items):
        # 如果收集阶段就有失败，提前退出
        if self.testsfailed and not self.config.get("continue_on_error"):
            raise InterruptedError(f"收集阶段{self.testsfailed}个错误")

        # 遍历执行所有测试
        for i, item in enumerate(session_items):
            nextitem = session_items[i + 1] if i + 1 < len(session_items) else None
            try:
                print(f"执行: {item}")
                # 实际调用: item.config.hook.pytest_runtest_protocol()
            except Exception as e:
                self.testsfailed += 1
                print(f"失败: {item} - {e}")

            if self.testsfailed >= 2:  # maxfail
                print("达到最大失败数，停止")
                break
        return True

# 模拟使用
runner = TestRunner()
runner.config = {"continue_on_error": False}
runner.pytest_runtestloop(["test1", "test2", "test3"])
````


**讲解**：

pytest 的测试循环体现了几个 **专业测试实践**：

1. **失败计数 + 提前退出** — `maxfail` 机制，失败到一定数量就停止
2. **nextitem 预判** — 知道下一个测试是什么，用于优化 setup/teardown
3. **异常捕获 + 计数** — 单个测试失败不影响其他测试执行

这正是 `unittest` 框架的核心设计思想。

### 1. requests — 异常体系设计

requests 设计了 **层级分明的异常体系**：

```python
# requests 的异常继承体系
RequestException(IOError)        # 基类

```

**借鉴**：自定义异常时，设计 **基类 → 子类** 的层级，方便调用方按需捕获。

---

### 2. flask — 装饰器与上下文管理

Flask 把装饰器用到了极致：路由注册、钩子函数、上下文管理。

**借鉴**：用装饰器分离 **业务逻辑** 和 **横切关注点**（日志、权限、缓存）。

---

### 3. django — 元类与ORM设计

Django ORM 用 **元类（metaclass）** 动态创建模型字段。

**借鉴**：高级特性（元类、描述符）在框架设计中很强大，但业务代码中慎用。

---

### 4. fastapi — 类型提示驱动

FastAPI 把 **类型提示** 用到了极致：参数校验、文档生成、序列化全靠类型提示。

**借鉴**：现代 Python 应该 **全面使用类型提示**，提升代码可维护性。

---

### 5. pandas — 复杂继承体系

pandas 的 DataFrame/Series 有复杂的继承和方法解析顺序（MRO）。

**借鉴**：多重继承要谨慎，优先用组合而非继承。

### 6. httpx — 同步异步统一

httpx 用 **同一套 API** 同时支持同步和异步请求。

**借鉴**：设计 API 时，尽量让 **接口一致**，降低学习成本。

---

### 7. rich — Protocol 协议

rich 大量使用 `Protocol` 定义 **鸭子类型接口**：

**借鉴**：用 `Protocol` 替代抽象基类，更灵活，不需要继承。

### 8. pytest — 钩子与插件系统

pytest 用 **钩子函数** 实现插件系统，所有功能都是插件。

**借鉴**：设计可扩展系统时，用 **钩子/事件机制**，让第三方能注入逻辑。

### 9. celery — 配置系统

celery 用 **类属性 + 配置字典** 管理复杂配置。

**借鉴**：配置项集中管理，提供合理默认值，支持运行时覆盖。

### 10. sqlalchemy — 描述符协议

SQLAlchemy 用 **描述符协议**（`__get__`/`__set__`）实现字段映射。

**借鉴**：描述符适合实现 **属性访问控制**（验证、缓存、触发事件）。

## 四、五条可借鉴的编程实践

从这 10 个项目中总结出的 **通用最佳实践**：

### 1. 全面使用类型提示

```python
# 差: 没有类型提示
def get_user(id):
    return db.query(id)

# 好: 有类型提示
def get_user(user_id: int) -> User | None:
    return db.query(user_id)
```

FastAPI、pytest、rich 全部大量使用类型提示。**这是现代 Python 的标配**。

### 2. 异常链 `raise ... from`

```python
# 差: 丢失原始异常信息
try:
    json.loads(data)
except json.JSONDecodeError:
    raise ValueError("无效JSON")

# 好: 保留异常链
try:
    json.loads(data)
except json.JSONDecodeError as e:
    raise ValueError("无效JSON") from e
```

Flask、FastAPI 都用 `from e` 保留异常链。

### 3. 文档字符串

```python
def calculate_discount(price, rate):
    """计算折扣价格。

    Args:
        price: 原价
        rate: 折扣率 (0-1)

    Returns:
        折扣后价格
    """
    return price * rate
```

所有顶级项目都有完善的文档字符串。

### 4. dataclass 替代手动定义

```python
# 差: 手动写 __init__
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# 好: 用 dataclass
from dataclasses import dataclass

@dataclass
class User:
    name: str
    age: int
```

FastAPI、pytest、rich 都大量使用 `@dataclass`。

### 5. 强制关键字参数

```python
# 差: 位置参数，调用时容易搞混
def create_user(name, age, city, email):
    ...

create_user("张三", 25, "北京", "a@b.com")  # 哪个是哪个？

# 好: 强制关键字
def create_user(name, *, age, city, email):
    ...

create_user("张三", age=25, city="北京", email="a@b.com")
```

FastAPI 的所有 API 都用 `*` 强制关键字参数。
